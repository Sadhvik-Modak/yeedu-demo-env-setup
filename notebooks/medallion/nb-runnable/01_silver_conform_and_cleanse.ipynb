```python
from IPython.display import Markdown, display
display(Markdown(r"""
# Step 1 · Bronze ➜ **Silver** — conform and cleanse

![Medallion architecture: bronze, silver, gold](data:image/png;base64,iVBORw0KGgoAAAANSUhEUgAABWIAAACmCAMAAABnVQbcAAAAYFBMVEX////w8PDo6Ojk5OTf39/1347Y3uTf3tny3IvwyJTg16/M0NXTwqDRrnmorK62m3CokB2amIaQhmdreolqeoioaSynaSx1cmllXU5QT0pBPjc0MzIrKSUaGRcREREAAAAh/2MWAAAuAklEQVR42u3dCXOjurIA4CSObVwqiSeYe7Xr/v9/+bpbC+Al8UwyWSatqjkmrDlgfWkaIT38jwsXLly4/KXywKeACxcuXJhYLly4cGFiuXDhwoULE8uFCxcuTCwXLly4MLFcuHDhwoWJ/Z3y3//8mPLfb3mBfs0/sPz6/hXr/35EYWKZWCaWiWVimdhPJVaefkSR35ZYcfhhRf0bxA77f7wMTCwTy8QysUwsE8vEMrFMLBPLxDKxTCwTy8QysUwsE8vEMrFMLBPLxDKxTCwTy8QysUwsE8vEMrFMLBPLxDKxTCwTy8QysUwsE8vEMrFMLBPLxDKxTCwTy8QysUwsE8vEMrFMLBPLxDKxTCwTy8QysUwsE8vEMrFMLBPLxDKxTCwTy8QysUwsE8vEMrFMLBPLxDKxTCwTy8QysUwsE8vEcmFimVgmlollYj+PWJehpGBhLZnrdCHL+uRxNs53OCfksawOZZnK5zv0f8HPCY4SDRNLQNmQotMwFbM+6BxorsuWPlPWql2Z+VAuUln7anHuoLJ6i5dxPujExN5TsEpFNxa+JhejnXAfNtu6Qqx18Y5d5brSmO/7NaYMu5+Y2E8hNnofoBoSpd57uMyIpC21NI2F3vEKsbAhlr9ErIyrH0bDxDafUg4OLpLeEqszOadyEkCsd1g0EBud8ynfVFTrPyDWrsWeFRN7X6lVikCVsUz7YUts8B4r3bsQK+P6h4mJ/TRiMWgdc56QUlmmRwgbM0Swo8tJwvxE6gaClrbxbcMrO3wnYqd4NoOJLb5lB/81KOuaWJFISpMxLM1tlyW2FY62uQXebxMbzoJiJva+OBKq1DAAtCBoyHEcTmNCXdfEAoKwRjq9B7FTPJvBxH4isae4EHtKME2KEqsW5kNoOy7ETjnKl4gdffKwQpw83N5LuDfysJ21Y0w4Uf56T7AbWG3aGEob1NmGVhpt2YgTBcutfTaophJbYuv8kPUlsYc5x7O9CBMSpg9KosCQkXMSBw1nf8YNrVnJq6NK5iBsiBb2jDc3B/HLp2AEJwruLvB3jj7NOOyxqhUg4fOM2P1QPjai+uSkt+gvVIipEmtimDbEYg0a9lSDiHTYT6lBnCj4ClGs3BCL/7CYHGF+mHLoxMpEnzeJjW4cPYSgMdgRVrZSOvjJRidPHvYyQrEQG0/wN32M633QBn22xSg2ubYyE1sL/O0xuui3IZYmRAYmrxEbzqH2SpikKrElkPUOdmKECmC1DU6JVZjqtDiECNtEAQfR6LFWKhom9u6SVnBClFIQw5nnxMIMf2YdVB7pI6zmvRwm4BWJndI4TGFDLNSgYYSKOeK6JuIBLNagExP76bnYNJ0qsdLAfcop94g1w/wIvE6N2FBsbbnY6XyHICKQPZ4ipXTpdh/2Dpe5BKOENGxESdwpr3472qDPpi1pKeUtmNgWf3rK4enzKFYkgFVjRqDnYhuxKlCEe/aQCuaL9rgr2APJiduAmOJg0/q30khwcRj8JWIFpSU8E3u3V5Qg6GnZimi4RiwENNttR4xZgc4ysfeOiA2W9rQmFvfqPW1xImL3csBDj0zsJ7cogDs/11oUQDGXxI7E7EjPwfx6w1zhG71c52JBZIpQHbVF8O5kPQXLtBItSYYaK4wrYjezC7E2UsKAiV2DZ/Ck2DNigVNN/w69RcHh0C5SrMdQvt79u2jUqkXBnEpGNaG8AmZYvzkihsUzQQqqErGHGZ+iRSb294ilPJm7l9iprjUlWmrrhI1EbJpo3TWxmEqg2RCXVGJLDWJiPz0Xa7OrLQoKrkuiIBCxJ58nj0tGfAB2ukwUjP1ZWCfWFFwviK1r9KTs5nlWn43EyuRG2hcTe55KxUYCW2Ixgk0YfG4SBdG5mL04f7QlZp/jr04squosfJSiL4gtGQrSuhBrIsTRhon9zUTBgLeMbp0oGF9IFJhUg15aG6LWMgGzgdiBVt7mYg1RXmoQESuTHYeBif0CxI4UraKAnhrBtsddnh53RVwhIrEyNhRfblFwRqxdETvFkvG1mJUd5RmxfTYSOyUKaZnYFa56ru1fz4kVKSlqOXCei1U5XWszoAw2nK3tYr0RuFIyCou4RuycaJkqxKZfqC4Te3/x9XHXlN0rj7vS+eMuikf3qUex4VYUa2i1UoNoqylRooCJ/XxiMQFbc7GpBKvYaEva0mgrErYY3xaAXyQ21ZiWSFxysY3YMZVw19Pm8rxVVp9tY916YmLXxKY8E3rA6JZYbM5Ft/AXj7sgKrqAWlCzgU7sHHQsc3Dp4RqxivYq6uMuOlJkYu8vWKVO+2GklgVwDzHCdLzWaAuWDeftvU63crHuglhseLAfZIGZcJ6Y2E9/3IXpAdNaFJhcYFu9ehBLnJvHsaX57PLqwXjeogDiXmpRQFlValEA6jZiZbQlSh3B3dHFc2L77CmNEnUdcYdMbC+/4KJgGrTnYhM92jLELj2k6o+7TGsXG/N8Rmx0SuikO7Eie0PvNeiDcvFqFAvrCqEjvvJg1SF4ISw+FGNif6NhbKYENuZkQ6lHjl49qA+Oy6sHcf1crBS4txuoRcEQvIQKWloUGHyKFS+IlWkapAOlsQaNeRpGC/OZ2M9+3OV7iwJsI0vtuGzoL9DGkjM4I7a/6LXdoTOptIs15TFYxuZandhKt6NHWWnjc9mgzZYe1DcxewilLRO78s5BNaXGq4XYegEpeeBKwrXPq4229EWqQHlslbB6gdaXLK2As+/U1SgWPnzGBl0HHZNQLiUtYIKJvT+OtT4FTy/Nbl+grQ+Oywu03sgrbxFAncCoVbpMzVyR2MFC7cCmA1ti91SDYB/4/ATCWKhrVIOY2E8ilruB+Y6Pu7gbmO9I7J8XbHpVcq/cDQwTy8QysUzsO1sXvZT29ddqmVgm9rKMqRUm9osVuJMtxTCxnx3F+hjdjQ64ZK9Bw18nVn2LwsSe0dgKE/vFiuiFif1kYl/2t5X9BxArvn5hYjlRwIkCJvZ7dsmtxMOXLwcmlollYplYJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCaWiWVimVgmlon9YcQ+va0wsTeJffoKZcfE3kHs0xcsTCwTy8QysUwsE8vEMrFMLBPLxDKxTCwTy8QysUwsE8vEMrFMLBP7lYj1uRYj6CM6hbOF8cmbY1vLwpIUyqJH7WLy864tUzbAqnAYXfdkYeacMx1YuBQ0E/vtiBU26N8+3cYysUwsE3tWjHOgrHNOC/pIZKMpWCbRiY20CLV0OTmfsn8sizRulnI6wlRyWPTDEeHGbQWsl/L8kcRqw8S+ndjZK8HEfn9i17XhvYrRTOxvJgrA1sfy8fzw8BwhClWA6fHh0ebQiTVFU/VwJDyfUlZlUazuGljcVjdRFWJxO5HT4wcSa5jYdyDWuD859UzsVyPW/AViHRP7JmLhFt8DmLZQqR/XxEJw63C1J5g8tkRBsXZ3fFgRqx4fCrER/xuaxn9ArLDeqRnq+2523iq4wF7r4A2EWMK48hmUd/hTwGtvQwjiSTn/wheBiW31VsFJU/UTz6Uxygcr6CxqPPl0zrVXYYbTbOEMC4fLz0rdDRLbr5MyZUIEYYKbd7AbuCj4qeCazs9M7N3EHunMwRnbGfhU8P3uE17bcGxf92V226LWhn7CNxcN5q7rEe7FKmvgcgu8nsqfVbN6HWGXnol9C7EaiE3nKFZiVU4PTzkHLR5Xydw0i6eSMgjLFkTsY847jHD1HxNrnIJrDMSaoI86QJWFL9NRePO0c3AvO3sB1x4q/pOF7wQtd/B3WwUjlJuZ2FeIVaGcNAUnV+GJM1DpdtaVKBbOshDGP0Ods0oIVNTAFC3fVta6GyS2Xadnb5Qy5fKoHayBUAvtAW5vhbBux8TeSyycT4VfZ5wAPAHNPoGk7trXfbO8bIEXdXXC19l2qMXK2aUePeNqDjbQsLtnmBDCz5tqVq7jc+Ao9k3E7kI2j/VZ1QWxAslUCbO0c3sUJgI94hLXiH2iiPcNxB7xch4hSKXrShXf2zKBV5u+QoIWHUX5if5ZSxVfMLEvE4vn6mnW5ROjHwQVJp7pTM8eauUzoEh/uug000nX4fzOse4GiF2ukzjCxrB6uTxQ3SmNIKD6+yPdejCx9xL7jN9tvJWjE23hOvUJqg3t677M3myxOuHrLC1e3aNa6hHuAIMTuL4wMeN/4JKdVTO4jkzsW4ilx13x+DKxD8/arR+FPWqL6KreokC/H7F02RdQFdwUwV9W+vaUig6LypfnODu4gynE7ujbAPOZ2BeJ3bVWA57+SAVdGCRrXcvjUVjzXOtorYbPS1JuvZteI/E6Qfxbsg2tjkJwpI4NhKIBE3sXsTvt6O68nGgQsE9Qbehf92V222IJOMoJx4s0tyjWavH0tNQj7emrgBcKszq+zt1WM9gLE/sWYrG448M6UaCoFdYqUVBnhzKn7cLm2FsUqE7sAwXG7s9zsStiSxGd2LnM8OXa7+Cwxydr6pegFM3Evkjsc7WR7hWopm6ILZUSKijVuSvE1s2fN8TW6yS8UbvdmtidNvA3EEKtsgYTezexswcMZ1+vE1yDPkG1oX/dl9lti3Jft5xwvBLtmaSCy2HVUo/KE053Ruy2mjGx75GLfaD2AfS46xFvH9bE4nxR9qizo1Wflar7uJIoeAjlcZd4M7Fwm6KwPC9RrKcZqlz7/gWgf5gHhMKJgvui2PJ5rNnUTqw16zp3hdjWLmBFbL9ONfJdEYuRk4YgydhycZjYO4nd0U2GuRnF9q/7anbdgs776oSfpWOV87tej+brUeymmjGx70UsuDo/PjyaHDctCgy6Oeewe3jYuRrFKsoYPMLca8QaarQV/7hdLF1XysXSl0bA16QRW/QVz6u6L0LLxVKIJPhx11252HmVi10TS7lYjIxeIbbvBnOx7TpRhdVrYqmOa6i09OBF8OOue4kl03bOX83FYm1oX/c+e7fZYnXC188oS0ZH9Hqkl1zsithtNWNi34vYllZNav3qQcj4EsEu5hz8kot1+EZYKrnYRuzsHCZ254djyi6+lop9qUWBc+JILQo0Ps+kCLUSC3ew4khPqEsUa47ClmfhSiivd8r4Zyb2tRYFWiGDpUWBfdoSC+e0tih4hdi2G9y6XSeYh5dgXqqmdYqaLRydeRb0NIWJvS9RQJXA4ENIr56oxUCfoNrQvu7L7GULqA2rE7553KUFbLRb1SNYzZ4nCrbVrN4nqiMT+1Zi6QXaZMX6BdoM9yM4vZsB2GiXp13AacIE7EKs62/SipDTW16gxRaXmrJEygRvjitid9oGapMSWstO+IvscELRsxbLUeyr7WK1DxR2amwXezwjFk8lNYl8jdi2m7J1vU4ar8AMSvdH0ibAWsfyZJJa4zKx9xErbPBK+CAEfuPpWWKboNrQvu5P6+VlC6oNywlfp3cwy4pXt9UjuGLeYkuvLbHbalZaKHjBxP4z3cBQW6w/fkXlDcSakIKp0zFeW2PKbSK9bt2YXFnXpzid7/80uhjdiLssa51CHrkbmBfLrzzjR/ljbl7zS9jk7tfOhPWmWR/c/RvH+S++3UVtsRC/PrEtN2bfWbCZXcnifmgfBcOUp+0EFOtvrR4nJvZdicW7S33epO8DiDVpGqdcDLS5ESvjHxJrU3RlGzuaPG33f5LJT1NIEvaUZZnBxL5MpgupEOvoSc6rfuVZvYFYre/e9neO89vEwg37UWHg2icubv6vzb6vHPEFE3xZ5GOJHVwssvaJl4mdxveAMkomtkWxxnn3p8K+gdiE+rlQvLNN1ukPiR2j9ESs9+ht2OwfZuBuJcg7hRLiWs/EvgyZE4VYb+7yS+ffSTxcEPtl+ijA2qB364nzriWuzr4zisX3pcXTBxM7+SFNm4mXiX2XIjMT+5c6Mwwh3EeslI2+k7dNVux/bDqNNiaPd/VJ2hSmSqwNCW/1T9KWe/7t3mAvrss9Qqi62n9bB1SdovWUmTCN2Dh5kFz6FOGQtDDgRtZdP9C3J7ZeoFej2MOhEBt+bZgKySOICs6Xg3WsVTE5dcAL5yD0hbkQZOqoklFZOXwTHLbAuFP75GcKj2EPnVgdolUlUXBj/ahnfGaBqYhIO8dEwXIgETL+QsLAt0P/y50Z1uv2B8TKYV9k7RNErBvhSoCDzsFPYx729H2XlCgAIOHO0sKhpIsBP8Hn5M2wfJaUAq0Fl23ELIRPtKq1cBuJ34e+Ryb2nYnF78Kdj7uCK5FrD17JRKB0tHRX72EiSyLWJohBE4gXrZTuSli7IlY2P4NbYYgB7hSRXzS4ExvsiHE07DTSgWSKEuW9eaBvT2yIdz3uKsRuUp8qaTWDiCrNQnkHxEbQz4USxYqIr+pHcdDAnRAZqLTor/O43AgVICK2QeFjuZ5eEAArEXtj/RhmISJOWKUc7Bx+ofWB1GGGQ+N7/iapf5tYrFh/9riry7oiFsKH0ceF2AG/7zYOhVg/DmMe9xIUls4Pe6g4wwRL2mchNk7DhKtOCdZI4zhGA3sODjeW+75HJvbdiQ3hPmINcifTeEYsbYohJzEIaKJ8ZKcNlc/xFrHBUhQ7LfuvmYScfeE8WoxRF2J9j3bTBH+9T6P3ExxF3jzQ9ycWkL2b2GRdCqau7C3e5evyCWHnAZ0jXvGfiRilwmYalgCZM64DoaVOh/I8SydRsgKNWIpmdSP22vqHiBPW09FgORF7fiBV/hC8kjH+B4gN4f2ITSeUdezE0q39MFZicb1oQVSMZGEti2tJ2T8LsfDDkG1JCww403jc84A7lPu+xx9BrA43yvH5TeXWbu8i1mBQenL+dE6sjZQwKLlYb5HYkcDDOTGYxT0/nRE7QbArfSG27L8ZO8Ug8UAmFkwbsWiyc3UPELzCfY6lX2hzoBeIvXEK7PMXKOrWBbqb2By10tGuZ9VP1BLsQxcLsaVZAPBLiVmUjxbRj3ULVaS0lVjaQjRir61/wAAWKRYQzqoaVq8O1LaPRr2ei3U3ToZ6/nLF3vhV349YzMUOQOkSxYZJ1hYFiOp+721ZuA9gbbIT5QvqZyEWYtY9aozrY6Ig51j3XKLYukcm9pOIdSTgmOQZsZJyrguxrhBbiiSAY5M1mzNiTzbnVMJfl86EjAYPBF8H2N9CrOnbwn+tPYVxDPi5PdCPJXb9MKs9mKqfEHZuiKXYFv2r8q3IFK0Do0Jsy8XSFteI7et3Yg/CxBw0EXt+IH0Qs8/xFxP7BmL3Ek/wuCW2joplAVCXkz31z0tiLcQte7Mmtu+REwXvnyiIdxCL2dZCIpWpE0sPt2QnNlAUCz+PWGor1yxvJApgoaTHXW3/Zd7YsI4YFgOgV4m1pzGe4KD5FKbbB/oXEgW/SWyBcZmVX4pizRViD6m+5X93FFvXX4iFFZUrOYHzAxH4yuSZEwV/SOyUcbcDPvnYRrGeah3FonKKbv15RmwylOFdE9v2yMR+yuOuqcaYEi+hjaPsxFKyYCJi17nYqTRDkFNJm94gdpyoSday/xIwUzOCVBIAU4RFW2JbLhaD54Dph/zCgf4BYu9/3CWsqklRUg1tM6Z8Ui52TewqRXpBrLOlnULxMLZcLG4xXyO2rb8kCsjRpM9ysYVYofH3c5Yfd/1GLrZmTElEkweJi7BV15pYQw+r5LCnO/4p9s8LYlHr/RDXxPY9MrGf0mgr2SUqXRIFAKNEXUcXMXj18tRaFExxPI3BQYRqpTTnwSVA7eEPrkSa4Q/nuNr/5PFhF/zkEuViQdd0OiO2tCjA6NnT07AYTjcO9FMabUEEiYGkEBEHXEi1caxOs8KnUxpbAgR32BIrkhX00P8KsQpwpGU+CuFib1FgDipcI7av34iFVYWYSxR7fiAtolNCJ82Ntq412hpHrAzDMlFaFNhhwBYFU5L7MeDjLvi+T2dR7ABrYSXYuzAOI4a19fMyig1eDhbhJmLhrlP2PTKx7z8U8evEyp5b3RArfTYnE7OXNltQNebeLnZ05QWuKebsz59D2Z5tsJlSucv+LW48hbIRHcjbc2JPo8/RUUta3AfoeutAP+XVg5qEE/hmbI799YMZfsDgdoYLg80MNsQeFJ5FbK56SSw2a8UWVtikNZneLnYOOahrxPb1e6JAw+Vo7WLPDqRxBrWP5UG+L0u9lHKZwOLslKhd7AAX2GKiAL/v7iwXux9MyB4RhRoWp6F/XhIrfUqTTEmWlxps8n2PH0Ks8d+N2NaH++1XqrHbCu4G5jVinbly2v5KubZj7gbmrxL72yMC137TxEcS+0O6gdkQa9Tb0YzPTCwTu+wUu0bUgon9CsQazcR+MrHh7cTuMhPLxK5qrXv3nraY2D8l1jGxH0vsI74UiMQejadeX3POETuBTWF+7AN1RbPDAbiEix7n6nj02SPFyiWH/cFaM2On3C5G80yjgNmHR3wxGzfKKvo/JlYY7DF2NYw7hETeiDWxyuAA8E9t/Heq0t4wsa8Q28+jst7Nz6vTZsogCM8woUNw4lJMmEnDdleTj9qVXRgz43zrSz+z9YJgs2dddqxKx6Z4KLiqTOz7EHv9tBOxwp/9ES3jdNEG7cIvl2WGq8LEvjOx2HvZHIBAFwQ2C3x4REt1UkKFuXXZrR+xzd/Dc3LHo4N1VXLiQeOA3xlH/zW4G6se8eGpcA5Hm8FREZ141Fk8HLPT4o+JtU4pTWNF12Hcd4764l+Ifcahi/B/Q+Nobzg8hsBO/C0T+yKx/TwKj6/Uu91y2hZisbd743bngW5drxMLF0hhZ85Pxln84milcACKfkFmV9YzcMwdLlHBCNiAiX0nYq+fdhzox51fPRoXhjboF75fFu1xYAom9l2JfaLRYT2w+bjD4QpMIfbxiMPD1NjTRBpqBoJUHMoLw1mFA3PhBHoK4D49mIQh75EG83okYlXGn5x9eH7LwDKlT+71MO7LKF0tisWuhJ9LRW81mYhgYl8itp9H7Y9np21F7O5K6oDG9FJrYqkraHQURzd52oldGee7X5BOLC7GHds6ugIT+z7EXj/txgKc514WYnGD5cK3y0L3hoaJfVdiRRnFEInF8WJyJRZu+fGN3jpMjKUVK6gPELSqjCkGmEODxqC1hhYph6/vH4lYTeOBzxGIVW8hlt4rNMu4FqVT93WiAIfVCG0UP7SWhovjRMHLxPbzSNzRRDttC7HuaTWGd0vi9fU6sTvt6P6zDhQNN6A0Hnu/IJ3YKmsZKhXmMLHvFMVePe3GrnrYNvMmin16Wi58H06oj2DLxL4fsSXWRGKdV489ip2j2kHQWtbxpoa7dUiuhdjH2ryzEquyFmBxIXamrQHa59fG+H6J2J1z6oiVuhM7nxMrvFG73ZpYx8S+Tmw/j/XVeLOctoVYrHt9DO9O7QWxs9dlzNlaWWdBNfYmsa3zCM3Evg+x1087DpbWuXR2TSz+0C98uyxHusKci/1LUWyZ8JVYik5Ni2JNjWLhroPKKoo1NGNXiLWuor2KYsPbiF0G19tGsatcrMbR+gRHsX8UxXqsYOV9+7MoVl9Esa1cRrHUrsM0YntQdJPYJ1/f8Wdi34/Yy9NuvHIXjQoWYvuF5yj2bxK7o0Rp9BR8PhxbooDu7UNcNemiXCyNMHt8WIglfXEOEUs/mUos5WtLLvYNxPaMYSf2IhdL4ZheE0spRM3E3pmLpQciYrecNqpyMxIbjiXrepb5q+vRCYfFFOfuXCPWtozebWJplZ3gx13vSOzlaYfZ4vzv44rY5cK3y8K52L/RosDF46MFYnfJPB4dZluDEQ/ePT+a8giL7v8fBLYoeAywzpxXUaxI8M/Gx0KsieJBG1j+jBkDbO2j6Yc3RbHmKKx1C7HH4J6xvVEnVgUaRX5evloK4KDRjZnY28T283h05lnMQSynDeMYmMDHXeZ4tOfPpOGOdFeGUofaKrBpgXPiaEzLLEDwtNOmEVwfRIoNscrTNXtmYt8xir047Tj7MiZ1Ru1og37h+2Whpl+eiX1fYp/xpWyNDbF89kJl/6BifMTRh9Qupl0JY1O0ghoaaJ+xqWsn9uFoIjbgKsQ+mpTnB8TYJPdwxPfxYZU3EVvaYKrSDrOkAWGO13b1uEtjo745mP7Vgr/PwXGLglfaxfbziE8UqTlxO23PJgRTcrFwbi/bxcJ6nmJbPPM4Ac56JaBulqprPCAMBvcLAsv1hlh6RGk5in1PYi9Pewlu3fPZ1fP+WF5KaBe+X5ZnQxeUif3wbmBET9r+lG5g3rmU3gpf7T/xdue14WJYMfsRI9C+8JL7u7wF9u7E4jiEvb8V5dL8cUSWXmK/1NtdX7AbmNLzq4xu2NPD8+ipRxd3MUIslkQLXesyaZ9qHzFjfvv4tBPtsw6POEwuxjJewvha11t/hViRzU6UBltM7G8XOzGxH0ZsCLN2tYPug0lXhstS4aZx4U1Eas3E3knsED3814VxHCePvWZdIXbqC2m9EXs5TMnuWzdZbycW9zmFeIKjhQSHm5IbPotYfEs2mEcm9o9KZGI/jNgywkzr3NBcM1PfhFSHb5so+F7EDj7gWo6cHHCMgivElk4Gg2vrUY/djrrgHpJ/D2LpN6Wuun0q4yag4J9E7M/rL7bd4FvsP3vyKdrS8zZ1wb3B0AbqAHYqg3OPJwkzyqiIOIi7POH9DhJLQ9SW0Q5xaETcZRlHIbUxFGmvbUGcTE5W0hhffkJicdR3miE9HPFjiP2y/cW+WCqx+CbNfMBuM6ygIBVTCL9oJsYMth5B/PI0hm1ZUruJTTiA9yHqGQIn6h02RNt/IxODwZ5j4+ywK9p6AOdwoAW36r2Wib0q2+AiDa5V6cSPm8TasCF2SiONamDPiS19N8NCaw2OhUjVCmdLmJAjXEUMnuvcNbE4ju0pT8tQC0zsxxKLY67LMRsctKB0kx0wKLVuNcTXVAaJgRUotPQRbjlwTe/gRifQQIpILAWiY+3j29Cg7hYXjNLWMWaQ2L4gRnOSNMRBgDVwmIMQcVXYPjopXWRib6Zjo1iiWJVnpZw/4DgFCseIqTOF8q51462VQhd70Psr4hgFMw4wMwuBiwLM6buFveBwh0AsSLscgLrtBsRV1kzsS8TaJPcrYl+MYjFcXRPrLPXFfUksGDrgCAZYYwfYTNJPcO8vhxAgMsHuudvc8yi2q4qDijOxH0wshauScKRbfpmNLff+bQAaU+/jZZrK4AiyjmSImkoMYwuxZUzEljCQdatYh/zqUWxf4MsgX5IGZgyhDiOO49nQjjMTe73kHNt42mSmquMe0LCwphCrE41mWEmmMWf9QqygEWB+wToRFba+Dr8YanhKe8L4lRb3AxCx82o9JvYqsTZXTwud041cLBI7WIwv18SOEGfKLC+JJRjTtIz5RTul2TZD0GzdMnf93A1zBC2gxQwwE/vRxPoyrEsqY8A4mBxxHO6+dYGPxqKlYQ0JSOzKEeLRukollkLcbnMZ1B0WRDtucrFtAUWpNtYDALplf32Co9ibo3q5ZmwhFjvbACJb44ISxbrV/mc65QuxBVT8bxs1Zk6HyjAhPtesb0kJtAO0wWdw6Fkm9jaxaar35a6N0L0itt3H14GX4rSsl5BQHM1wivtObFu/KOltWTKlMvxXmU2jJcL8PnfVoiBjm4WfTexfKr9F7JhNGZgbolRrT0n2obxw2TLGVyL7fBgpirXxjFjYakxLekGeDK5hQi5Z2kJsX9BGni2pBZC1Dv8V647fnIv92uVNudjWfKpoSjfzSrTb9+KuSTTQVsms6jI67W1iTT33JcilPXVi2wH+IrHfqNxBLNzMF8VKSwFavxObp3WLgjI6bGtRUKJUvw9mITZ3YutuaElh1KQtsX3u0qKAQugfnij4CsQ637KoWYKffrJ2NYzi1EaqNQHZK1FnuBLFntLYc7hpGbr7NNKQh5XYvuCC2CmVL9p7RbH/JLFKFzQXTYlbfRbFIpTa17XSr7rJObFiFcXSy/zVZCJ2bov7AZjY+1sU0POuJQFwK1FgKGu7ThSggm3k78v2AWdRbLgaxYbNFti2QTaozU983PUViPW2tSTwmBGwNiyp2FNqyVQJBAKvY8uUkrXSy4VYa+O4YvcUQVLak7ed2L6gEVtysTFUa+XpvXKx/ySxc8LNnF80pcSpBfro6ZYxJbRthnYxY1znYueK7jKCN64qxKrBgmvE9gMwsXe3iz2VdrGvEjtEf04szAr7a8SO2JhrKktWudgVsVdzsaSr/cmNtr4CsSjjhI23TvRkf4wZ07NVywkWjsGVO3cINuEyAazUvMCPo6es6lgD4dQTDAHotfgsLbpRTmmJYpcFLZz1SZ4cHtd5TFXAqsmfZOAo9mprguiU+rXOCZioDjMKqNOsNCzB8TzmpIVyNYoNXgg45YKW0BzbWxSYHgsLoaPubzQcSosCWtwOwMTe/3bXiBnY14mF9c4ed+HDq+kqsV4OpsW3A1Yj/GlDbJ+7blFAug7+0189+NnEnnAk95OHcHKk2DH7lg+g+3yXojuV23cJ9/ljyAHiWcDW5+wkPQarrx6kzmIf1B3WzmHVLrYv6MTiHGsCPQnLlIEdcR3PxF51ysH51KsWBQIunjl4sHVOOQKbwuUZlMyptXNVLoG4MQlaUpp9pRTmw4rYg4Zz3+SE9YLpudh+ACb2fmIp6ryD2L2PwxmxROCVRlsQ+WCj2bJEuvLy7YbYPnfTaIsC5U9/gfbnEvsCvr+/g1VDhC/QDcw//LjrLxdKGMzx497u+peI/UulPO76xt3AMLEXdIXf3mIM5sTE/gPEYppB/+k7XEwsE8vE/p3i73xAxcR+dWIPxpc8AhPLxDKxX4fYr9ZfLBP7bfqLZWL//f5imVgmlollYplYJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCaWiWVimVgm9rOIVeLrFyaWiWVimdjvSux3KEwsE8vEMrHfklguTCwTy8QysUzsP0Ysdj744sCzLww4y8TeS+yM3eGH1ru2siEu4xmoNth3Lzr/dbgvD8rErkqcPp3I0WcpXZo2L4T9xu/FxH4lYiccSnFiYv8msUqp2VH3hcJlo9UcgvgkYhX27PVLMLEvvOg63qOYjH+RWO/HwaRRyuns94qSif1KxI73EEvDEzCxf5VYWs+gpr8KqSL4PydWqDcQqwMnCt4lUTD9TWKDXYY+2PafyMR+JWJlXlnpU8S+DcsIsqXr7mimkijAG9nWR2ymfrQlrI4detM4MlPtlDZOFpc66jzW2jEm6vObBn5nYl8l9hDt4RDcllb61D75uYw2E34RsQbH5qaJkCjDEGeXl+TCWszsBXU0SyN2KYhQfQoGVnUWTdfYxyj8pKMIOLr3L7jWMx7UWhXLrvvRmdh+Q44DyMJ3fSgds2LHsNLFYKljVwkncqKRC6f9QF9+WiP60qFs6yYWq5IZaPxYqEo0Hgx2GTvsoSrSgIkWLtBZvCxtwGUDDsKGA5hSoiDCf5M94QSOwue2h2FivwixYzbjiMMcdGKnBP9CzcXWgbmwu+5xAkVlshK+U+fEQrA7ZV9G5rLRyZOngQ5GadPIxL5KrImHPgDiIZlOrM5GKOxMG4c0ADMFbKEFjVlgsE/CBBAiuYdLYkX+JZRfEzsnrRT2YmiDU0KBpwrHsNHIKf4iZThGIBbldTiYeDs6E7smFu7UxzwOEceeNXGQyUON8ABkcpLGTzSRbujlgD/A+pMccaDuoQyXuJeRqtL+BJGNtLAD4tRbHEzhNEaKUt149ovgIIs0/EyLYonYOA0n3CCWoWc3h2Fiv0wUK8nZFbGhDfa1JpYGQ5zayLRg65ZYHDgRuzvE7S2N9p1rkkFKJvZVYue8ygsE24ktgxcmUeYZWJvG2oKwVZRhEYHI6NZPq5YuCyOlFtZRrKpbWBoirA0AhmMj0vKFWFGSEv3oTOyaWBQMXKMRYQG8KQ4lmkVQaQKJ7SNt0frE3pjKzi1uMFnkuSxBNU+wnS8Dd532Nl38GoMsDG+JdTVorcRuDsPEfh6xdcznNrbM6BMN892JJTinM2K9K5/O1Z+2xNo6aCIR24a8dTjuF+di30AsjXQo2gSyl+tEWb2PD9OaJ2Ap+yEfN4mCw4xXGuS1vo5IQyvrXFfsxPqKbz86E7smtseckhKgztUUKY31MlRi63ixsa6PDlq7HYambAh7ohxBGkr4idYut/pTmxospgfOiTV1TNpK7OYwTOznESvplmMcZUsUyNJ+oBGbrxJr19ReEGuuEXsyISfDxL5OrMVEwXyRKBAVTV2yCABqn1Dd006sUEpjC4WySwp3N8SaqAUFt9VQo2oy4gax/ehM7DViETh8sOXLWTojlmLcvUl1fbB4SDW96u36kwLdEX8Y6vmeVsSaVGHEhgSXUew5sevDMLFfJVHgGoj3RbGVWHsXsfBh88TEvkasQFX9lcddydDLg2IVxeqymcq6LlqPEbNKFFxGsSW10IgtWYCXo9h2dCb2GrFgmzc4wuyIRV6NYkMbKdGbMQ7bKNa3KBb+4Wb7ZOueLh9Ylb2E14hdH4aJ/SrE+pZ5xewpDU1LiVV3noulBKvf5mJJVn+T2Gkd/jKxt4ml5Kgu4aLwq0ZbjnIGojY3cC0tQLlYXYc6vEFsz8USobgz2n9sxFKUa18kth+dib1GLEwha5RS3cthS+wqF0vETqHfwFMLgtENtuViIRimhC6FtcO1EWpbjPoqsavDMLFfhVgbx9NkQcYRiJQeJgy2KoiV2JZOlfgntrcoSOXpmAunk4k3iY1ulFPiKPaVVw+ULzkCB7fuSoe4evVAAaTKwRyD4WikhgTqQC0K5giLMfK9QazK5qACbAbxLr7WoA7BC2FRczLUwPbzhlhstbAhth+dib1G7D7S86kh2kGavIpipzTKIXh5MpSuJWKHlFq71VqV9rVxDj0so/3JNA0040oUm+AgHscLv0WszJPcHIaJ/VRi18WmbE8+Y1ya/VjaxWbfEgVTrO1icaB3S0/HMjbKImJB5GRuJwrGkKmhLBP70gu02bdGV7NLOZrNC7TCliaqwsTsMFEAWAZDDWSVS9GdRbGbfYccKTIGmj157VLSIiZBhgq47ubgs+7Ewq7nDbH96EzsVWItNS7YD4aawy7EYm3aS5eTH9stPuYTlpAUq9JA78JCVaJsQXlvQGJb2qtR7B7qUpBjDjeJ3dvkt4dhYr8KsdwNzOcRe/F2Vn6tpT+mBgDNu/enDx9U+O2ul2X7mO4NXj8ME8vE/uietmx65fHSjG8cRMPEfiti5fhqdPlRh2FimdgfTSzcvqeX78yNT2E+MLHfitgpe/kBxN5zGCaWieX+Yrm/2H8vUcD9xTKxTCwTy8QysVyYWCaWiWVimVgmlollYplYJpaJZWKZWCaWieXCxDKxTCwTy8QysUwsE8vEMrFMLBPLxDKxTCwXJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCaWiWVimVgmlollYplYJpaJZWKZWCb2SxH7Y8p3JfYHln+C2J9QmFgmlollYplYJpYLFy5cvm1hYrlw4cKFieXChQsXJpYLFy5cuDCxXLhw4cLEcuHChQsTy4ULFy5cLsv/Aw2CVCFhjKB+AAAAAElFTkSuQmCC)

Silver is the **trusted, modelled layer**. Bronze answered *"what arrived?"*; silver
answers *"what is true?"*. Every table below is deliberate, and every transformation
is one a real pipeline has to do:

| what silver does | why it matters |
|---|---|
| **Conforms** three channels onto one schema | store, catalog and web name the same column three different ways |
| **Cleanses** — trims, upper-cases, normalises countries | `USA` / `United States` / `us` are one country |
| **Deduplicates** CDC replays | the same order line arrived twice; last write wins |
| **Quarantines** rather than deletes | a rejected row goes to `dq_rejects_sales` with a reason, not to `/dev/null` |
| **Records** a data-quality ledger | `dq_metrics` is the audit evidence that any of this happened |

### The star schema this builds

![Silver star schema around fct_sales](data:image/png;base64,iVBORw0KGgoAAAANSUhEUgAAAxkAAAF/CAMAAAASD/SjAAAAYFBMVEXY3uT//wD134703o3e17LJz9XFwKGvtryhrLehrLa1rIObo6uwomOokB2njxuCi5SFhHRtfYtreolqeolqeodpeohtcHBRUEo6OTUzMzMyMjIvLy4jIyIXFxYREREAAADBe3BWAAAn6UlEQVR42u2dC3uruK6GOeHxsAmsQmZ1bINt+v//5fENQtK0gYaLTT/tPas0pamF9VqSL0ryAYFAPkuCRwCBgAwIBGRAICADAgEZEAjIgED2JuO/v3vKv3Pb/TdGuczV8t/gVTqMRv99Scal+N9+kv87/wn/Lzq5zLajXTtlghSzWf9bxdI3VzJIsqPMJ+PfJDqp5pNBwtYom09GEUvfgAyQATJABsgAGSADZIAMkAEyQAbIABkgA2SADJABMkAGyAAZIANkgAyQATJABsgAGSADZIAMkAEyQAbIABkgA2SADJABMkAGyAAZIANkgAyQATJABsgAGSAjRDI4TSr17R1FR6Igg3TFN6owFhEZzzslNjKMEcnLtCZo7YcO3ZeM7H06GRUNmoyvVSkqTUcVDxlPOuWhUYuwyajz2MiYoNQgNGwyvhcRERk/aUzgZEyV/cnIuGSZd9yyop2khHXNjfVQ2b5XWilCuWwpSVjXdVmSMdlQEhIZRStp5qOpx6roaEq3XeoHxiWv7AuVkoxclKQhkTGhUwqhjAY2PnSd00rtDt+1flVCmPtpJTN9kRVS62g6Ur+ttjP9jATdmozeiHQ0ddeqQSMmJMtH2l87VBaKGqX0t/ZGZ3nFBAt8gQzBiLZx1wlCj6dVx4vkXY1jJ1mQqtFKXVSR22dqWp3p1meMB0RGpi4kZ56Mx6oYO9KGowcvSnKjCROUZIq/k/yFoWl5Mp53SqGq4qLbPJDBeEGoyhJqfAbjGbno0atSLCNNwwgx5FNVkUrpaLPjVbYxGYMRGTJuWjV4e41FTiW5an/tUA10ljRCa6h/nnUXUnA2fF2JDOvgiPKdwIz3MmbfjZ5cY5pPzbCU2XHHkfFudHjq6rYkw9pE0ZPxUJWeDGdP2tKYNI7b8D01M9yCjAmdYi2CVlcyZG0smtinUNg79Y8qc0ENUeY+RV3/Zd0zB7k8GYMRGTJuWnVFwxvUoP21QzvtSnLzj3mfyvQZyYevK5FxkXaM8p1gmm8s58biVeUfvo5BlI1FDBlOqYaGQ4ZtUT839ViVngx18bfa37H/tDQcMqZ2SjKKpvQgW/Tjg/t9KvUP9Nd34YBwpqXvzZ76x+XJGIzIkjFq1eiPMmFDwUH7a4ca47vYeToLNrMur/+6EhlUegP/shOIecGCTGVF7C8YMnhnhYVDBp9MBnFt76pAyZjaKWMyyIV3NmQS/e9TdWuDhdc624GMqxF9SQaxSYa+b9D+2qGddThW9A8Lqjpzb/91f59ho0I6+IzcSBaYz8gm+QxJbeNJxD6jviNDS06NXQkT6Voy2lsbzLrKaR2mzzDBrSXo1mdkPRmVtK23JJCKy/HXNciwMZ8JPL/uhMZaj1bKci88GTYITJ55s03zDGn1mUKGCwWzJFAyJnSKyzO89ej4nVTEaXLNMzi7tUFirZOQZA8yBiP6kgxrUFVXXbW/dqj5hdwqpZufF+67/utKZBDFCWnkd51A9TcXqZVqeEYY02jTJifEzOqYyZBgyCj0w8zbp2QI7YBzWSU5E6GSMaFTKlXl+j/9NUvyRpOhYxFSSfOCmcbRPUV1WnFrg5XIEzepsz0ZgxF9SUahdSzMlNSg/bVDO9d5xGp40UG9zkmGr2vN2uZcyYp91wmEyo6ZEDFj+lYiFckbleuXm45XAfmMpGo7kT8lo5Jm5o9KxYL1GRM6RavRmTVLYmJtE6XkOvEz3UFY9256SvEiubfBgpnFm13IGIzo6wycio5pnumg/bVDO6cy69wUl+oUza5f11wDX02woxA7CgPrG5ABMkDGKmQw6YVGT8YyqoRAxtqabE9G3mskSSxkkEGiJ2MZVUIgY21NdvAZm/cNoilEU4imQAbIABkgA2SADJABMkAGyAAZIANkgAyQATJABsgAGSADZByNjMwfa9F3v3PFh4PnGZfDaXpmdvwS2tjNaruRkTXytmWFuejbS5g/d//54u6RMdmrmVGuGrs1TZqjMXx06p5QIZ+W4dmIDGY7yG6p042yO+mGC9uMvmM+X4RIhjGz98dq3XTva9q8Tgaxp0LcgTBa0P6kcKZYcfF7EQrlzu/SgslsNzIK6cgYWpYrmhfSny0kgldVwx9d3MpF1XleSa6fTq1Ypd/K6GQO6VfUvuqtURQ57fIwyGD+qBiTle4CMrqwrVCm34pHFyGScbHb6OkjtW6690VtFoqmiCz8uUTW+N5oEn/aSpuY9RkWk4buRoYs3JmvoWXjU23ua9YfHb65uBU3HuUy7+sFEFMnwY0BpOfMH51+tktpIzL6CkxOGVNbo7+4amSKuXy+CJEMexSf8Qdq3Rrei9osRAYdTnX3G+dtQ93BAMpN8YrM2goVu5GR+dOQQ8vsjpu+xpc7rNeflLy5uOuY4QXqi9Vk2dAv1XBQLBub5M5k9MORG6h0Dw0XV8Ma6h+ML4Ikw3absaNPat0Y3qvaLENGNpzQM1WJbPPdQXxzTFL/0JDRF6DYMQO3ZIxaZs1bjsdVTcPni7vQRPZ1J8YVizwZQ/0B/1iqIMiwdXMSe77bPYXhIhnGhrzLPl+Em4GbM7Cf1LoxvFe1WYaM3mWwrrcFf4xMuuprhgzXZDoV3RXJKEamnPem3PQcfL64G7J04ucq4o1jpf5ajb0EEyQMMhiT5uShG1b14DRcDHGkcemfL4Il411lD9S6MbxXtVmGDOnNK8upR+Nqf5UtDmfyJOEOHodERjGUUXtGxqgib3Zhomu+JKPfGE1YQ5IgyOiEzll1CnQYMi4yf6RWeGRUo+qQF3c9ODWTmzsyiD1vPLV6/SbRVKUut1nq19HUXbFHU6rii2iKuylswluShEFGDwBt+miqiTqaYrJ4qFZ40ZSzHleepfAtsJmrJfZaeC0nyYXvTcbQMld35TYg1D/7fHGfyPekDBl4NZBhgsXMljYiDSdJSGRo03DDkqnNqUbQu1FWv/T5Ikwybqb+x2rdGN6r2ixChque4eYALj5csiDX0i136BE2S4o8mVHAfj0y+pbdJsiFIs63fb64XRVxNBFbV9iNCHyIrLJrmvE8lNqMjMyugmnHTuz0iA59h4uruzSG9fkiRDIqmX+h1o3hvarNImT4LNYsblV2jYXZBRf9zXs/42zzjOK6HrM9GcTUyzYLQ0PLGPf1Bu0zbnlRtLZKzqeL+7mpPC+4UaRQTGssVO5W+vJ3NQRQWXeZUIhxGzKI7DuG9kti1wu7JHZxi5KfLwIkw1Y+N275kVp0ZHgvarMEGf1UJWGqc3U6bWW4thtSVJdnsE7SycHe4mT44qz02jLpXjE1tHPbPOU3hdxf3Gp7YbLzu0bM7hC3L8HuDmEVuYa/VpogoqmMSt8XVHU8G1+41U7ZCbeN4tNFeGRcdyN9VstqMxjea9qss6OQJy/LpjsKWZbsIvvvKMwW/vzBfXcULqnNKmQULC4yCE9+Kxm0OhIZS2qDXeh7Cnah43wGyAAZIANkgAyQATJABsgAGSADZIAMkAEyQAbIABkgA2SADJABMkAGyAAZIANkgAyQATJABsgAGSADZIAMkAEyQAbIABkgA2SADJABMkAGyAAZIANkgAyQATK+JqPId5T5ZPzNo5PLfDKKsDUq5pNRxdI3Axn//X1J2PtLvz6bjL+rCaWrvfVlPv+h67anRotq89+XZLworfg4iDTtx2GlaY6kTbtqT4EMkAEyQAbIABmbk6HwvKHbxtqIKMhoYT3QDT4DZEA3+IyJrWxgPRHkGeJQ2nzEQIYAGcjAD9VRyWLtPMpw1CCaQjC1JBniKA9dHpmMA+kiVh6Kk+We+lE89bGC8aO6jNXD9+QDaPyacOpAmq0foiR48L8nBccE4k5kHASNw5JxHJexxRxbsvCzP0KMftRE4yjIy00UWZYMHf4d4OkfNdE4SrS7zapMsni72wM8eriMcFPvrfhOVrCrFjYElxE7GCuQYQyrBRkBanUEuLdTIllHgciT2PaAOXj8uOskdsM4N1mpG5qog/UD5uDRq9Q027KdrPXGmm+JyANucLmhdmMFkhWViTimOtz0VNwuQ2eum5tSsq59RcvGwZLwmI/PmPRih+Ynq8PewpQQS70UeezTFcn6msXJBg75BDE+Nbv1Q7L+n2iaKCeqDpRqNG20XOyHdLKNlcW4MH4YrxEl48IMqHs2INlM0TY6l34QMiIkXLbN7tu2ky17KLKpqmNk4fGBEYahJJuamla5gVFBhydYBGEjycZ/T0Q1V3WAbcNR+T09dAZjHcke5hZPPi6wa3hDiPWgGc6OomSPP2rZEJF0V8xgxOPzTMod1rNOdvq7IpZVjphzjVjabgbKNrT9p8l+f9qmHOF7jni9RhRg2AnaED99JdnZ6vZezzlwrhEB0nayMtDBMdn/4QS/Wz1Kr9EEfzzGYhHuqJME8oiC9vwCW1uWH230eBj0gJiE0QzZBL3O0TQtwFjuadqRMPQnmgT0vELOyOOaogo4/mvjmHcJiQz32IINPSNKNkSoltc0EZ1ISMLr1kDhiMZrBDrbI6LxFoGS4Z5haAuiEXmNNsxhJfBJljjI+LDlrsPbQCIiqBMUYCQlmjbCwzmhkuGeaGjLQDL4YS+4vZptpGedgyZjSDoCGqkDP14SVvMidRZRkPHh9tUElJJ/P/69rSN/3iYG8+E4DDMH30R9vCWJoZFuWiMUNMRXTPx5+/PH/res6Hf0hDwzxlCcq51Bib32ShJJO+0QFETWIT+NhAaKNZB4BMh3g3QgAV0TfSn8qMjwQ1EQs7k3NvjmuNhEvoEjiNRbtpGtWRyFjA9/UHj3R3+1wj6GMv+UZXpaS9KyvPLxAI4QxosmhiMFhyXDD48ajl1japf1XN1FeT5tIGnZuw79d++c6f75dnuwj+NJomz1/rs12/aKRXraTtLBcYzbsqdN2tqzge8o/z1kDN57N5sYuNgSCydln3EMo8TO/vuYH2uYxNx4ex5wl3kQD0Z52kVcWOXyjf3mpISl4pifEB07GUNOvjEd3l/sxIULqlzev1fdHPPUm2P6isOQ4YbNTeMqt3yxIxfXhOPtbY/HHfWuj99EhnftG/WWdxjpaW9xM1XboiHddoTm4/iSHEiXZoucfNcE41EufjuBu3K63Rw8hDooGR++rteasfefQBzGKKTaJKJqj59YHJsM14nrLTsZj/FWnsKRLdBwkxy/iYqDkuGHuDUCq7fpDiOlspuOEOvoS6sba85uGDesPn6dJIfVzBVMXXI+920GGKeyU3S0Z0SwlchwifgqXsNVyPsV6favIsPnjAuen7FgTDXYuhuzcO5WI8PP364Qkv6CqdlfS4Yb+Rbq4llgsE5LeTozqXh5ou676zjPpGRlf3H2ZKRM2G/0b+mLOp2Xhy/obZvf7Ct+DxlalFjgZMcsMLTJd6I+p7LjmpFSf9fW51GkJRlV3fl0Vh1lnUwtGSnvmAbpfEqVYrXo6lleY6E0XLmFofbj10vyWxR99WTHPDB8NFV3PNUOg9/FVtRY/bk8n0qugVDamxgySoNIqW80/53S+nzaGI0mqhqCIGPxtENsAYZjgfcD/y0ZdTcOloQng1ogOqmTEu0zzrNzjdfQaA94xAJkzJ9wmXvuyU5LpbPJEI/JSG0Wwv85nXnnMhBDhn1RS3qqpfnxrGWT9O2FVMN9+hUiqN9Nxoedz23nneKfD0bvM+gjMk6nf0qmU4xTq+2/lIPP0N+U9sxsWmo4+Px1jZ9Q4WIokAAyfEztlgIn+g4zIM/cEWJZYCbPqD/lGSVLTRrOUjt91Y3zjLTWr9DShVXzV8N/4CuQbYOMR8PlpBkr4zLmbgmxLJzd3NTZrvuVowyc19REWlxf6BtYaeemRCcod4jQmj9ZAXmExtuMVKMV8BUg40lS/owOG0udfkKGTSRMwpDy8XpGSpV5OT2VXLFS32PJOJ11hCVMZl6Lzl3Mz8LhK0DGkoGV3Un67bxUKLtrn6Qaz52GOz7fSvQ7yJhBh/wKjCW219bMy2p7dZ85DV/8RqG/QcY8OtqHZzt+EkvtdFzjm6lbsdL2Y5Dxa+mIJZbqncYDNOzKNiIokPFqUn67BdHk35GAYUqK3JPhfEUDXwEylhD7oR2+ROf8pYydk/BP2TaoABlLiiupJOLJMm5nbn2RDyxYgIxV6GhYVC7DOw33IW7wFSBjRYnLZbgknOHYEchYHYy3eCampq1pQEDGr3QZM/aIQEDGS2SUcZEBpwEyNnIZE4Kps69z4AbtmitO05OrjpCeRhf9HUK680uS222DZat4uWwODgEZAQRTqeRlyaW3fSap31Qu65K36fXiugndkPGPZGWtqDZlRUvalQinQMbRgqnaQHH2tp3ar6x1L6SqvF70HkYxQwY1Z/bKUnNltqszjnAKZMQExtuUUfo8EDGUm+KnWhkvwdn1wt/NaW3IkHVfpM0WnuqcT2GMujJV2g9pb5Kac4Fn41fSm5AN4RTIiGVmquyuVpvW2kPUakDkNHYK2sMYMjRJVEiWeibO/tcZZ2Wq3cg/GoT0H/MiF6UBRr92Nq9NCqfegAbICGRmysVETnhnQieHChfXC3ejgUYaFERdlooNZLi/xIyboAaAVNUnUWuCalOEpLbopJM+RxbhFMgIhox/Wna12XPJtFWnxiHUsr1e+HDp5MkwZUWoSu/I4P4eG3/pi5KXQuNxTmU7tQYVyAAZ6wdT0xbAS8lubzS5xVl2HTOWPlycTi518NGUi8Huo6khJ9H/lvJEaarp0Q7nTGUnSszbgoyI0oyzuoZSLuAp+9jHDv/DxWkot1abYMne5xFR6QMy0i7ViTgv3S+nJeuQgoOMAMiYeGYpFeNQyhp63aUWEZNVDBcu0tJCpUGF+wlfy4EHx7065Bknrt9Iuw3tK87l6Wbu99tw6g0pOMgIIM2ola0teD6ZAT7lQl/rfNyt//HT9UKbeD89ZRGiZW2qe5aqvq70WTK0D0rdPBQV+vtSaDz07edzPclnnFb4PA2QAZlPho+QTPBTm00fqpN2d4jo7C6R6wXnIzJOOk+3EZUpYyvL04gMnXZ3bu3ClUW3hTxLk62USMFBRhBkzKtIfirrJ/ursKkQZByDjJlnM/j3JNGN9u0iBQcZgZERZJ0ECMhYHIy3E8gAGZDPZPwBGSADAjIgIANkQEAGyICADJABMkAG5qZABsjYCw2sZ4AMyGcy3t4iJQN7bUFG/D7j3J1/9Huyxr4pkLEbGd9udPr5B+7Jf0AGyDguGeKnZKQdyAAZUecZj6dtS67MKb6uM7UG/TfaULk5X8SkoJ9CsJR19qB4yYTSjqa0hznO1BXttGT4AlOmoqegfe2p4e+Ny06llEvrrGRd8/vj5wMZyDNAxro+48/DxKBOS3tIu7QH8tKzOa8qOS3Ts+LnM+N35ppyZg71mTJSZUmVKTGlfQYzZ/9M1TVt8n2BqVPdlalF7KYwyajs1ImpMi3NoVdpqvJIhqkpkLELGemjw63mlHbpyLCHuE09BNkf676rVtiXSzjr4T31P7VkpPaUH7VkDAe/7dvdkzEuO+UKK5iTf9IASFUKMkBGIJNT547VQ71Oa6jmlKukQ50D4Y97s/Gxb18zXQdSPRm0dVGVJqMvMGWPy5ryCrdvMCo75bISE2/ZP3gfePVk4Bg4yNghBS+Z6tjZkuFMeSCD9yfCT6dRKVsqhzqG57T3GSnnpa0bYmydD2SMa9wObzAqO1X6qjxnl4GXj9J3JOAgY59Ew8RSXKZjMkrvM7ivIvI5/BrCME+Gq77ziYwbn3GdyxrKTulfs/KNz0BdW5CxT6JRlm6u9S7PMIZaC1srJH2UZ/Cz9R2lJ8O96KOpayFb+ijPGJWdchWnTI5ik+9HeQbSDJCxTzhVq9JVgxKsTEszUdWauSlqa7LR9EzvI5xUaFfChL5X38/0jf909Tm1NaV0Km3IGApMUfVgbmpcdkrDd7YzYycp6/T8aG4KwRTI2CecSqnqlKkGVUodUtWic+sZ1P2s7T5/sliq8w/zC7XsWEo7M/XKTU0pcS671ubUfYGplOl3+xwijcpOmSxHugUUv0qCz1wCGYGEUyuK/ZCaYS4LG21BRmSzU6tJrXRwJinKsIGMWGen1hLKlahfcVMIpkDGJmREd3oJYICMYzoNkAEykIOvdNC11SLRdSBjRRFNG5fTcFlG22g2GtABMlaRpjX/a+NyGuVbH0sZNnTzFToSZKyARZ+Ex5Nl3NRGaCwdokF/goyFQij9PxFjpvHn82ZCYV3HVR0IyPiJtFdfEeH01FdrGQ6OFnSAjBd8xWf7MWSUsbiML6dsWxdaISsHGfOoaG9DqE9obB5PyfonM7bfr2U4OFrkHSDj5yHUDRk/iqfq1z6usp7tptK3KYt8wtAB3wEyZqbbH186jdmGytgOSca05W9DB1YDQcZ3VBgsJtxpRuOJ8VRfRorbw9qu2pRUvLbHAFlnq1GlTLiP+/Y/5iVtfcWqtObK3qOjqesvvJpkPHSUNrICHSDjU2JhgoqJIbdxGhNTjb6M1FDe4MT5Oa3Nwb2Ol6k5/p22skxpX8GTiXPKOPcVq2qlf92cebVk+F+YuMY3e8MUfAfIuDcIM2DOmcT8MzkL78tIXQt/WNdhSyHUruSOBSf1tXdsUYRUWjL6o0uUezLqUY2eJ67KsPt/P3sY8B0g48kk1DOvMYkMX0ZqIKNWvrLOUC3EvcL4CBxbWs2eJNfRlK0KaskoR6V1lkoyvqYD6x2/lgxhN9q1P/rdiWgMZaTuqk1pGgZDp65klPRk/DMmg+pAy/7KLDJeA2M8ZwU6fh0Z1lm8skliGhpXa77zGWLsM0ZVqW59hivOOZeMBcAY0YF9Vr+HjFd8xUw0hjJSj/IMb+jOTfyT+nLOJulQPRkuBZlHRvpnweNKNglD3vELyJBuy8cC4+DbFDSGMlIn1paprzbVz031hs54mtb9Ijfn//i5KRNNteY7ptIZZCwKhhU3kLQg48DOwi5uLxUdvE2ZvO3LSJ3OrSp9nefOf1BGb+im2FrbTzmdecfrIc84c6UzDanSyWSUf1Y54NpYaUHG8cQNfIvGzG8z1jWmr12nLy2Ym3WMtU5+i9+clB+TDHcoYYUBz7DxtuzGWybs+t4r22vXLIkgml9KR3JAKtZcurJeY1G3kTK3neSne6U2KHwufuPhp4OR4Qa4NftwlYjq54VC/mz00d9+A3sLMuJ0Fu0GC7lvy7uNn3OxaXEpaXdg/pbI6iBkmM2BTbORw7dsvO18zM8HUn82H35+S95xBDJsGLxlZ3m3Ue7Jxduf3aoRuu0ER6cjdjJ2mlh8c3DsxEbq1jD2/MyxdpvQFWT8OC1sGyH3+MMGDXvUL90n7w7hs/hE8+JWNJCx1mTJTof93RYs7zY2dhyp+5uaiyAKHQi7cnRIOqIkQ4g9S2AMQcTbFY50Uyxc5h3KFKqvRiRBxu6Zxa4Rrrghsg+q9D/2o4jXyyzKHgrz/953hdQpR9tlFRcZzZP6N5tknp9ycQ/H+vJ2l3a3jQxryGoOlHZERIbdIrjzk38Ywg1B1epkfC5aG9YwrdrjBFaRkOGWLPY2g28M8U/vOt5W8hRvjydpw9uvIdz+dZCxWWax/7N+Hry8rSR/vvOkIUa9TRN9YJWET0UoI1Cg9WBFmBao5tTzAhmzh55wFlrD3Weqgp0TEk0b72JHuGSIoHY9B73/OuSh2R0pVyBjGZFtYFvWAp+qD/vcRJyfTZCE+CDbwMpJiuC3zgVvd0HMLUZNhquuCbM7HrwfdpU8nu25QZHRBvlpcpHMzUexN8NtXgcZ8wa9MDf8x7MbKJJIvonk2FMSytMKdCgREZ3OiSaOlzFsXQ+ADBFwchbXjEpMm10XKTV8ZDJE0OftRWT7quPaBy6Crg+aAIsDhCfRtlgGsiUuKDLCT8RiPIkT4YdfNGHOVyV7dWDbhm54cW74iXITXxOgOST7PIcIdmHGWjIm0jOnkz+B+rhkRLJNIN6zN9EWnxVB1ZXemIw2lq1lMZ/YjLhSgU0+5a8jI6Itl3GfR4u6ZHkoH4SWbNhd8Wwni/2Mf+QHTYOYrNqIDBHVofn4PyaiiRxtsX94kWw1BkTUVQeomXSAz/TeeyJ3AzJiK153iArfh/h0pF3ZSFbHIrZa8m3zcQQ5RilNsV9ymqyuWHQmdQgwotsM+c1ItY8NJat2ToQD12EKsx7E9+3GxnpkxPk5uQf69NLmMGjYqc2jkCEirU/XqMOY06E+onj7uDxZq1daWBMCw4VjkDZ6MkSsMa5oj2VLH4cSsWl8nqxDN2wpCKfxcTARG7qNZAWy4zWvg32Mb9MeDY2P7aapEvTGkcOPD6ARChlxb4A+miU18uOIbMgIyYh7OuRwcXnbHJCMD7lJXLIoGU3cY9ThXMZBydimpxYlo4UdBZZoHJSMLY6WLUlG7CtLx1oZM6KOSsYGEz1JTI1deyBShzOgVhyUDBkRGdFPEcoDDrDtYZ3G6ua2HBnRD09HDMrb9qhkrG5vy5HRwIp+Yzh+WHe4GBnxr7c2IANkrEDGASrRIJqCaiDjkF7vV5EhYiGjQZ6BaAo+A2SADJABMkDGbybj/e+LQumr7/A+t+3//V1UKFv2/f7dWZ9lemUBvf79u4a8s7/ryB0Zf6v/7SzFZe4Tf7/8L2CpZuuzfx9M0uvvXL0uUejVq/ffPRn/S3YWMp+MKglYsvlkFEkEUswnI0vikQJkgAyQATJABsgAGSADZIAMkAEyQAbIABkgA2SADJABMkAGyAAZIANkgAyQATJABsgAGSADZIAMkAEyQAbIABkgA2SADJABMkAGyAAZIANkgAyQATJABsgAGQGTQXmSyEtcZBQdSRib05BKffWTWe+zEhkL98HXym5MhtYr6/Jnre1+2mzz2FYmo37W+oRVoZFRVMsYi3mfydqtSMbzPpimKNWtfA+HDPJOppIxu9kbkPFcRHBkLDmMigDIWEgoDSiamqTXQMZ8XVcjgzDJKu/JK5kJxbNCSja2uoLJhpKk6zqpv+FKmJ/KimnTrLjk1fZkUNm+Vy6ayrqMdaLKGsVG/WTCEtKZGElcksy1X5Ohlb1pLqGtNL5Cv4/Tboo+y5PxvA+yruCdoPZC6BsvjWqM+cuKdpIS1jWm1b5vmNYls8PA9baLuu3TTcjo9TLRFGOVaYJuxw22FVecGjK8DqbZsmbdnfY3fdyb4PDYNB6SFUuTwZqMUOF7Rf9l0jSMkHHzs+5CCq6trNMtz1VFMvON4DTX4zYluaBbk1HJglRNTwbXj13qR8ZGowfVLSyEMHhkmf5hZn6o1Sty2o26k/GCUJXZPMNoN0mf5cmY0gdcx46qMhdVpkeGghRSN1hoT1d1vEjeFbn2jbU1lYxvuxDyTK/lyej1smQISjLF30neFeOerEjltNZdo3WwZMgiyZpb7Ud9PKg5PDaqX6lUsSwZxBhE0vheMVZj2neTklbS3Jc72zEtSnJ9n2B95lopsjEZ9qnRngxr0cVtsGTaTO2Tlsm7NOOLvqMydxE1imRlbZ47GciYpM/iZEzoAz06ucedddToUFnHmdheIJ31Jdm1bzwZt7c9jTsWJ2PQy5JhrIjz5HaaweqYZb5rTB8ZMthd6Hvbx72a18em6BT1ZpKR2yG09+Qm2nsXd38m61hlH4FpiX3Ypk12BFKXxCm0LRm2EUM0VXgtxrEqMfZR6EBJOw9nYZqmymYmfDQYMUmLvoOsdlP0WZyMSX1QOCO3F5n9hcKMT9T3i2n00DeejLvbaLsxGYNeLprqQWjpOPWjFfGdaZtuyaC+2TfaD33cqzm8fW6nvqoneedcMornvZIUVHXM+QzHadLV9mGTzkm1LRmuEcV3ZCRNRRTRmaiOXblrJPM3jIdicuGd6QdPxjR9FiejmEKG6XyqnLruF/I7Mq5948m4u21zMornZCQ6W1U6B7RaD2S8D1g9IGNQc3j7wndbtoLPYN+SodusM1PXA92Nz5A0N0KC8xk6lCr0I+PE3mQbmT3wGeYBUK1M7zMm6bOSz2DTfUbunQG59Rndnc+4u20nn8G+JUOrVkl2R8b3PqO78Rn27asJ3fazPEN+1yt54bUc8ozBQbvx99kMxfJ5hn3K35JRCO0wdKDdaF2Eb2RlBl6iLiPgiVPiJs94ps9KeYb8ngza5xnFTZ4xJmPom895xi5kDHp9E02Zd7iIOzKM4u8q+YKMXs3h7Z2e5NnU2+y5Kakz/m975SIrkjFh8rhc31KR3Fimfdi5rJKcia3JoPpJXeS3ZBDV6te56QViZkXMDJSZ9knGc1NEspxU0vkMo90kfVaYm3raB1nXFNoaqsF5FHZK55aMoW9oo9VS97dtTsag19dksKaw8563ZOheyAT9ioxBzeGxVSJP7Ozp0usZin6fZ+g0Q1HTLCmJBrwTZnHAtTyj8mYdYRsyCJUd+z7P0NkFcQiZ25vOrFJUspBmhnPkDnUOwr2vsNpN0WeV9YwnfaADBt4p2odV+gGoxvi+GzKGvskblRsTu71tezJ6vb4mg7Cuk+/3eQbVr96t5oz7uFdzeGxJwSYs12BH4dqyx47C5zuPXpdgdhSus4MPZIAMkLEqGbX0wuIhI+/bLL/1rMNdReBkXPvgWGQ8sS1PxsTe3JwMMkhEPmNam19UbTsyluuDsMggM/sJ0RSiqT0FJ5dABsgAGSADZIAMkAEyQAbIABkgA2SADJABMkAGyAAZIANkREcGYX5PKhXSbrHLmJQsi48Mp0fBpd2EFjcZ0h7QKexeOreDbrg4BBmDjS2p1sJkZKJxZDBZFUyShAiW5/xH6/S7klEoo0euaF68vs9lbzIUdQd0iOBVZQ5ADxeHIGOwsUXVWpgMRt2JKbfpWdLkYqDIuiIyMvTDNnq4ygcdiZuM4YC61SSzJzJ+3CsBkjHY2KJqLe0z/PF6V0fD7IG3Tq6rIiODcquH3XpTxE+G31bojh4Ker04SDTlbWxRtRbPwB0Z7rTLRfZjVRYXGZnKsqHNVEYeTQ2quKPs2n6Gi+Nk4JU/rbuYWmuRwV1rnYn/LFLfkQz9iAdzyrsqejIoV+YIYtNbTnM4MqyNNZGRQZqfNXU/MipzftWTUSx2HmY3MvKOFfm7Bvy4ZDgbi4KMZoim8p/O7exGBpFD1TKdMC12UGzn9QzWHDea8jYWQzTlaiiaNmbqpwnRbmRUXV+DzVWHPAYZOl2ifYZKD5aBexujEWTgxNbRldWPQ6ldfYaZ/a+6KtOPfMG1w93IKOziF+NJoYjroOHiEGQMNraoWguTYS0qN7WH/UpfrQpX6C+uWVv/gBmfUMQuggyc5bakYtLyomhthZr+4ghkXG1sSbUWJsOFIbbqkeq4sS33Co2TDLetYqGIar9oKmeuPJbZu6P87hB/cQQyrja2pFqLR1MLOknsKMSOwgDUAxkgA2SADJABMkAGyAAZIANkgAyQATJABsgAGSADZIAMkAEyQAbIABkgA2SADJABMkAGyAAZIANkgAyQATJABsgAGSADZIAMkAEyQAbIABkgA2SADJABMkAGyAAZIANkgAyQETgZVb6zFPPJuOQBSzWfjCqPQKr5ZBR5PFLdk/H+d3d5n/vE//sbtPx7MH1+rNe/f6OSOzIgEMhYQAYEAjIgEJABgYAMCARkQCAgAwLZS/4ftJHM2ykrni0AAAAASUVORK5CYII=)

The dimensions are built **whole** — all 500,000 customers, all 102,000 products. Only
the sales facts are sliced for the demo (see the note in the next cell), so every join
and every business question is still answered against complete reference data.
"""))
```

```python
# =====================================================================
# MEDALLION / RETAIL  ·  STEP 1 — BRONZE  ➜  SILVER
# ---------------------------------------------------------------------
# Silver is the trusted, modelled layer. Everything here is deliberate:
#
#   * TYPE      — explicit decimals/dates/ints, never inferred
#   * DEDUP     — CDC replays and MDM re-sends collapsed on natural keys
#   * CONFORM   — three sales channels (store / catalog / web) unified into
#                 ONE fact grain; three return feeds into another
#   * STANDARD  — trimmed names, upper-cased states, one spelling per country
#   * ENRICH    — customer + demographics + household + income band -> one
#                 dimension; business-readable column names throughout
#   * QUARANTINE— rows that fail the contract go to a reject table, they do
#                 not silently vanish and they do not pollute the facts
#   * MEASURE   — every rule writes a row into silver_retail.dq_metrics
#
# Produces 20 tables: 15 conformed dimensions, 3 facts, 2 quality tables.
# =====================================================================
import time
from pyspark.sql import SparkSession, functions as F, Window as W
from pyspark.sql.types import StructType, StructField, StringType, LongType, TimestampType

# ---------------------------------------------------------------------
# DEMO CONSOLE — live, readable progress
# ---------------------------------------------------------------------
# Every helper below flushes. Spark steps here take tens of seconds, and
# without an explicit flush a notebook's output can arrive in one burst at the
# very end — which reads as "nothing is happening" to an audience watching.
# Each step also announces itself BEFORE the work starts, so the pause is
# narrated rather than silent, and reports rows plus throughput when it lands.
import builtins as _bi
import functools as _ft
print = _ft.partial(_bi.print, flush=True)

_W = 78


def fmt(n):
    return "{:,}".format(int(n))


def rule(ch="─"):
    print(ch * _W)


def banner(title, subtitle=""):
    print()
    print("╔" + "═" * (_W - 2) + "╗")
    print("║ " + title.ljust(_W - 3) + "║")
    if subtitle:
        print("║ " + subtitle.ljust(_W - 3) + "║")
    print("╚" + "═" * (_W - 2) + "╝")


def section(num, title):
    print()
    print("▸ %s  %s" % (num, title))
    rule()


def working(msg):
    print("   … %s" % msg)


def ok(name, rows, secs, note=""):
    rate = "%12s rows/s" % fmt(rows / secs) if secs > 0.05 else " " * 18
    print("   ✓ %-28s %14s rows %7.1fs %s%s" % (name, fmt(rows), secs, rate, note))


def skipped(name, rows, what="already built"):
    print("   ~ %-28s %14s rows       —   %s" % (name, fmt(rows), what))


def metric(table, rule_name, value):
    print("       · %-20s %-30s %12s" % (table, rule_name, fmt(value)))


def finished(t0, what):
    print()
    rule("═")
    print("  %s COMPLETE in %.1fs" % (what.upper(), time.time() - t0))
    rule("═")

BRONZE_DB = "bronze_retail"
SILVER_DB = "silver_retail"

spark = SparkSession.builder.appName("medallion-silver").getOrCreate()
# 8 cores, and at demo scale every shuffle is a few hundred thousand rows. The
# default 200 partitions turns a 2-second write into 20 seconds of pure task
# scheduling; 16 keeps every core busy with none of that overhead.
spark.conf.set("spark.sql.shuffle.partitions", "16")
spark.conf.set("spark.sql.adaptive.enabled", "true")
spark.conf.set("spark.sql.adaptive.coalescePartitions.enabled", "true")

# ---------------------------------------------------------------------
# DEMO SCALE
# ---------------------------------------------------------------------
# Bronze is landed at the full TPC-DS SF10 — 24 tables, ~192M rows — because a
# raw layer should look like a real raw layer. Silver deliberately processes
# only a slice of the *facts*: this cluster is 8 cores sharing a 2.85 GB heap
# with no local disk, so a full-volume shuffle spills to network storage and
# does not finish. Dimensions are always processed whole, so every join, every
# mart and every business question below is still answered against complete
# reference data.
#
# The slice is taken on a hash of the natural key, not with .sample(). That
# matters: sampling rows at random would split a CDC replay from its original
# and quietly destroy the duplicate-detection story this layer exists to show.
# Hashing the key keeps every replay of a kept order together.
#
# Set to 100 for the full dataset on a cluster that can take it.
KEEP_PCT = 5

def demo_slice(df, *key_cols):
    """Keep a deterministic ~KEEP_PCT% of natural keys, replays included."""
    if KEEP_PCT >= 100:
        return df
    return df.filter(F.abs(F.hash(*[F.col(c) for c in key_cols])) % 100 < KEEP_PCT)
spark.conf.set("spark.sql.sources.partitionOverwriteMode", "dynamic")
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", str(64 * 1024 * 1024))
# A Parquet writer buffers a whole row group in the heap. The default 128 MB,
# times one writer per concurrent task, does not fit a 2.85 GB driver, so
# trade a little compression for headroom.
spark.sparkContext._jsc.hadoopConfiguration().setInt("parquet.block.size", 32 * 1024 * 1024)

spark.sql(f"CREATE DATABASE IF NOT EXISTS {SILVER_DB} "
          f"COMMENT 'Conformed, cleansed, deduplicated retail model — the trusted layer'")

BATCH_ID = (spark.table(f"{BRONZE_DB}.store_sales")
                 .select("_batch_id").limit(1).collect()[0]["_batch_id"])
T_RUN = time.time()
banner("SILVER  ·  CONFORM & CLEANSE",
       "typed, deduplicated, quarantined — the trusted layer")
print("   silver database   : %s" % SILVER_DB)
print("   processing batch  : %s" % BATCH_ID)
print("   shuffle partitions: %s" % spark.conf.get("spark.sql.shuffle.partitions"))
rule()

DQ = []            # (table, rule, scope, value) — flushed to dq_metrics at the end
STAGE_TIMES = {}


def dq(table, rule, scope, value):
    DQ.append((table, rule, scope, int(value)))
    metric(table, rule, value)


def bronze(name):
    """Read a bronze table and drop the landing-zone audit columns."""
    df = spark.table(f"{BRONZE_DB}.{name}")
    return df.drop("_source_system", "_source_file", "_batch_id", "_ingest_ts")


def publish(df, name, partition_by=None, files=None, comment=""):
    t0 = time.time()
    if already_done(name):
        n = spark.table(f"{SILVER_DB}.{name}").count()
        skipped(name, n, "already written for this batch")
        return n
    working("building %s …" % name)
    if files:
        df = df.repartition(files)
    if partition_by:
        # Sorted input means each task fills one Hive partition at a time and
        # keeps a single Parquet writer open instead of one per year.
        df = df.sortWithinPartitions(partition_by)
    writer = (df.withColumn("_silver_batch_id", F.lit(BATCH_ID))
                .withColumn("_silver_ts", F.current_timestamp())
                .write.mode("overwrite").format("parquet").option("compression", "snappy"))
    if partition_by:
        writer = writer.partitionBy(partition_by)
    writer.saveAsTable(f"{SILVER_DB}.{name}")
    if comment:
        spark.sql(f"ALTER TABLE {SILVER_DB}.{name} SET TBLPROPERTIES ('comment' = '{comment}')")
    n = spark.table(f"{SILVER_DB}.{name}").count()
    STAGE_TIMES[name] = time.time() - t0
    ok(name, n, STAGE_TIMES[name])
    return n


# Runs are capped at roughly 22 minutes on this platform, so every step is
# idempotent: a table already written for THIS bronze batch is left alone.
# Re-running a notebook therefore resumes instead of starting over.
# Set False to force a clean rebuild — needed after changing KEEP_PCT, since the
# staged tables from a previous scale would otherwise be skipped and silently
# reused. A full rebuild at demo scale takes about two minutes.
SKIP_EXISTING = True


def already_done(name, stamped=True):
    if not SKIP_EXISTING:
        return False
    try:
        t = spark.table(f"{SILVER_DB}.{name}")
        if stamped and "_silver_batch_id" in t.columns:
            got = t.select("_silver_batch_id").limit(1).collect()
            return bool(got) and got[0][0] == BATCH_ID
        return bool(t.limit(1).collect())
    except Exception:
        return False


def stage(df, name):
    """Materialise an intermediate result to disk. On a small driver heap this
    beats .cache() — the heap cannot hold tens of millions of rows, and a
    spilled cache is slower to re-read than plain Parquet."""
    if already_done(name, stamped=False):
        print("   ~ staged %-26s              —   already present" % name)
        return
    t0 = time.time()
    working("staging %s to disk …" % name)
    df.write.mode("overwrite").format("parquet").saveAsTable(f"{SILVER_DB}.{name}")
    print("   · staged %-26s %26.1fs" % (name, time.time() - t0))


def dedup(df, keys, order_col=None):
    """Keep one row per natural key — last one wins, as a CDC merge would.

    A row_number() window, deliberately. The obvious alternative — max() over a
    struct of every column — was tried and is far worse here: the natural key is
    close to unique, so the aggregate builds one wide buffer per key for tens of
    millions of keys and collapses into a spilling sort-aggregate. The window
    only has to sort inside each shuffle partition, which is bounded."""
    order = F.col(order_col).desc() if order_col else F.lit(1)
    w = W.partitionBy(*[F.col(k) for k in keys]).orderBy(order)
    return (df.withColumn("_rn", F.row_number().over(w))
              .filter(F.col("_rn") == 1).drop("_rn"))

print("helpers ready")
```

```python
# ---------------------------------------------------------------------
# 1.1  CALENDAR & CLOCK — dim_date, dim_time
#      Every other silver table hangs off these, so they go first.
# ---------------------------------------------------------------------
section("1.1", "CALENDAR AND CLOCK")
d = bronze("date_dim")
dim_date = (d
    .select(
        F.col("d_date_sk").cast("int").alias("date_sk"),
        F.col("d_date_id").alias("date_id"),
        F.col("d_date").cast("date").alias("calendar_date"),
        F.col("d_year").cast("int").alias("calendar_year"),
        F.col("d_moy").cast("int").alias("month_of_year"),
        F.col("d_dom").cast("int").alias("day_of_month"),
        F.col("d_dow").cast("int").alias("day_of_week"),
        F.col("d_qoy").cast("int").alias("quarter_of_year"),
        F.col("d_day_name").alias("day_name"),
        F.col("d_quarter_name").alias("quarter_name"),
        F.col("d_month_seq").cast("int").alias("month_seq"),
        F.col("d_week_seq").cast("int").alias("week_seq"),
        F.col("d_fy_year").cast("int").alias("fiscal_year"),
        (F.col("d_holiday") == "Y").alias("is_holiday"),
        (F.col("d_weekend") == "Y").alias("is_weekend"))
    .withColumn("year_month", F.date_format("calendar_date", "yyyy-MM"))
    .withColumn("month_start", F.trunc("calendar_date", "month"))
    .withColumn("is_business_day", ~F.col("is_weekend") & ~F.col("is_holiday"))
    .filter(F.col("date_sk").isNotNull()))
publish(dedup(dim_date, ["date_sk"]), "dim_date", files=1,
        comment="Conformed calendar with fiscal and business-day attributes")

dim_time = (bronze("time_dim").select(
        F.col("t_time_sk").cast("int").alias("time_sk"),
        F.col("t_hour").cast("int").alias("hour_of_day"),
        F.col("t_minute").cast("int").alias("minute_of_hour"),
        F.col("t_am_pm").alias("am_pm"),
        F.col("t_shift").alias("shift_name"),
        F.col("t_meal_time").alias("meal_time"))
    .withColumn("day_part",
        F.when(F.col("hour_of_day").between(5, 11), "MORNING")
         .when(F.col("hour_of_day").between(12, 16), "AFTERNOON")
         .when(F.col("hour_of_day").between(17, 21), "EVENING")
         .otherwise("OVERNIGHT")))
publish(dedup(dim_time, ["time_sk"]), "dim_time", files=1,
        comment="Time-of-day dimension with retail day-parts")
```

```python
from IPython.display import Markdown, display
display(Markdown(r"""
### 1.2 · Address — the first real cleansing job

The first table where bronze is visibly dirty. Watch the `dq ·` lines in the output:
**5,005 duplicate keys, 61,840 lower-case state codes, 12,590 null ZIPs and four
different spellings of the same two countries** go in — and two clean country
spellings come out.
"""))
```

```python
# ---------------------------------------------------------------------
# 1.2  ADDRESS — the first real cleansing job
#      Bronze holds 'ca', 'CA', 'United States', 'USA', 'us', null zips
#      and duplicate address rows from the MDM re-send. Fix all of it,
#      and count what we fixed.
# ---------------------------------------------------------------------
section("1.2", "ADDRESS — THE FIRST REAL CLEANSING JOB")
a = bronze("customer_address")
dq("dim_address", "bronze rows", "in", a.count())
dq("dim_address", "duplicate address keys", "in",
   a.groupBy("ca_address_sk").count().filter("count > 1").count())
dq("dim_address", "lower-case state codes", "in",
   a.filter(F.col("ca_state") != F.upper(F.col("ca_state"))).count())
dq("dim_address", "null zip", "in", a.filter(F.col("ca_zip").isNull()).count())
dq("dim_address", "distinct country spellings", "in",
   a.select("ca_country").distinct().count())

dim_address = (dedup(a, ["ca_address_sk"], "ca_address_id")
    .select(
        F.col("ca_address_sk").cast("int").alias("address_sk"),
        F.col("ca_address_id").alias("address_id"),
        F.trim(F.concat_ws(" ", F.col("ca_street_number"), F.col("ca_street_name"),
                           F.col("ca_street_type"), F.col("ca_suite_number"))).alias("street_address"),
        F.initcap(F.trim(F.col("ca_city"))).alias("city"),
        F.initcap(F.trim(F.col("ca_county"))).alias("county"),
        # standardise: state codes are always upper case, two characters
        F.upper(F.trim(F.col("ca_state"))).alias("state_code"),
        # standardise: one spelling of the country, unknown flagged not blanked
        F.when(F.upper(F.trim(F.col("ca_country"))).isin("USA", "US", "UNITED STATES"),
               F.lit("United States"))
         .when(F.col("ca_country").isNull(), F.lit("Unknown"))
         .otherwise(F.initcap(F.trim(F.col("ca_country")))).alias("country"),
        F.lpad(F.coalesce(F.trim(F.col("ca_zip")), F.lit("")), 5, "0").alias("postal_code"),
        F.col("ca_gmt_offset").cast("decimal(5,2)").alias("gmt_offset"),
        F.upper(F.trim(F.col("ca_location_type"))).alias("location_type"))
    .withColumn("is_zip_missing", F.col("postal_code") == "00000")
    .withColumn("region",
        F.when(F.col("state_code").isin("CT","ME","MA","NH","RI","VT","NJ","NY","PA"), "NORTHEAST")
         .when(F.col("state_code").isin("IL","IN","MI","OH","WI","IA","KS","MN","MO","NE","ND","SD"), "MIDWEST")
         .when(F.col("state_code").isin("DE","FL","GA","MD","NC","SC","VA","DC","WV","AL","KY","MS","TN","AR","LA","OK","TX"), "SOUTH")
         .when(F.col("state_code").isin("AZ","CO","ID","MT","NV","NM","UT","WY","AK","CA","HI","OR","WA"), "WEST")
         .otherwise("UNKNOWN")))
n = publish(dim_address, "dim_address", files=1,
            comment="Standardised postal address with US census region")
dq("dim_address", "rows published", "out", n)
dq("dim_address", "country spellings after", "out",
   spark.table(f"{SILVER_DB}.dim_address").select("country").distinct().count())
```

```python
# ---------------------------------------------------------------------
# 1.3  CUSTOMER — the flagship conforming job
#      Four bronze tables (customer, customer_demographics,
#      household_demographics, income_band) plus the standardised address
#      collapse into one dimension a business user can actually query.
# ---------------------------------------------------------------------
section("1.3", "CUSTOMER — THE FLAGSHIP CONFORMING JOB")
c   = bronze("customer")
cd  = bronze("customer_demographics")
hd  = bronze("household_demographics")
ib  = bronze("income_band")
addr = spark.table(f"{SILVER_DB}.dim_address")

dq("dim_customer", "bronze rows", "in", c.count())
dq("dim_customer", "duplicate customer keys", "in",
   c.groupBy("c_customer_sk").count().filter("count > 1").count())
dq("dim_customer", "missing email", "in", c.filter(F.col("c_email_address").isNull()).count())
dq("dim_customer", "untrimmed first name", "in",
   c.filter(F.col("c_first_name") != F.trim(F.col("c_first_name"))).count())

c1 = dedup(c, ["c_customer_sk"], "c_customer_id")

dim_customer = (c1.alias("c")
    .join(cd.alias("cd"), F.col("c.c_current_cdemo_sk") == F.col("cd.cd_demo_sk"), "left")
    .join(hd.alias("hd"), F.col("c.c_current_hdemo_sk") == F.col("hd.hd_demo_sk"), "left")
    .join(ib.alias("ib"), F.col("hd.hd_income_band_sk") == F.col("ib.ib_income_band_sk"), "left")
    .join(addr.alias("a"), F.col("c.c_current_addr_sk") == F.col("a.address_sk"), "left")
    .select(
        F.col("c.c_customer_sk").cast("int").alias("customer_sk"),
        F.col("c.c_customer_id").alias("customer_id"),
        # name hygiene: trim, then proper-case whatever the CRM shouted or whispered
        F.initcap(F.trim(F.col("c.c_first_name"))).alias("first_name"),
        F.initcap(F.trim(F.col("c.c_last_name"))).alias("last_name"),
        F.trim(F.col("c.c_salutation")).alias("salutation"),
        F.lower(F.trim(F.col("c.c_email_address"))).alias("email_address"),
        (F.col("c.c_preferred_cust_flag") == "Y").alias("is_preferred"),
        F.col("c.c_birth_year").cast("int").alias("birth_year"),
        F.col("c.c_birth_country").alias("birth_country"),
        F.col("c.c_first_sales_date_sk").cast("int").alias("first_sales_date_sk"),
        F.col("c.c_current_addr_sk").cast("int").alias("address_sk"),
        F.col("cd.cd_gender").alias("gender"),
        F.col("cd.cd_marital_status").alias("marital_status"),
        F.col("cd.cd_education_status").alias("education_status"),
        F.col("cd.cd_credit_rating").alias("credit_rating"),
        F.col("cd.cd_purchase_estimate").cast("int").alias("purchase_estimate"),
        F.col("cd.cd_dep_count").cast("int").alias("dependent_count"),
        F.col("hd.hd_buy_potential").alias("buy_potential"),
        F.col("hd.hd_vehicle_count").cast("int").alias("vehicle_count"),
        F.col("ib.ib_lower_bound").cast("int").alias("income_lower_bound"),
        F.col("ib.ib_upper_bound").cast("int").alias("income_upper_bound"),
        F.col("a.city").alias("city"), F.col("a.state_code").alias("state_code"),
        F.col("a.postal_code").alias("postal_code"), F.col("a.country").alias("country"),
        F.col("a.region").alias("region"))
    # derived business attributes — the reason a dimension exists
    .withColumn("full_name", F.concat_ws(" ", F.col("first_name"), F.col("last_name")))
    .withColumn("has_valid_email",
                F.col("email_address").isNotNull() & F.col("email_address").contains("@"))
    .withColumn("age_years",
                F.when(F.col("birth_year").isNotNull(), F.lit(2003) - F.col("birth_year")))
    .withColumn("age_band",
        F.when(F.col("age_years") < 25, "18-24")
         .when(F.col("age_years") < 35, "25-34")
         .when(F.col("age_years") < 45, "35-44")
         .when(F.col("age_years") < 55, "45-54")
         .when(F.col("age_years") < 65, "55-64")
         .when(F.col("age_years") >= 65, "65+")
         .otherwise("UNKNOWN"))
    .withColumn("income_band_label",
        F.when(F.col("income_lower_bound").isNull(), "UNKNOWN")
         .otherwise(F.concat(F.lit("$"), (F.col("income_lower_bound") / 1000).cast("int"),
                             F.lit("k-$"), (F.col("income_upper_bound") / 1000).cast("int"), F.lit("k")))))
n = publish(dim_customer, "dim_customer", files=2,
            comment="Customer 360 dimension: CRM + demographics + household + income + address")
dq("dim_customer", "rows published", "out", n)
dq("dim_customer", "valid email", "out",
   spark.table(f"{SILVER_DB}.dim_customer").filter("has_valid_email").count())
dq("dim_customer", "demographics matched", "out",
   spark.table(f"{SILVER_DB}.dim_customer").filter("gender is not null").count())
```

```python
from IPython.display import Markdown, display
display(Markdown(r"""
### 1.4 · Product, with SCD-2 history

`dim_item` is a **slowly changing dimension, type 2**: the source carries
`rec_start_date` / `rec_end_date`, so a product that changed price or category has
*multiple* rows, only one of which is current.

That is why 102,000 rows contain only **51,000 current products**. Getting this wrong
is the classic lakehouse bug — join to a SCD-2 dimension without filtering
`is_current`, and every sale is counted once per historical version of its product.
"""))
```

```python
# ---------------------------------------------------------------------
# 1.4  PRODUCT — dim_item, with the SCD-2 history the source already carries
# ---------------------------------------------------------------------
section("1.4", "PRODUCT, WITH SCD-2 HISTORY")
i = bronze("item")
dq("dim_item", "bronze rows", "in", i.count())
dq("dim_item", "duplicate item keys", "in",
   i.groupBy("i_item_sk").count().filter("count > 1").count())
dq("dim_item", "missing price", "in", i.filter(F.col("i_current_price").isNull()).count())
dq("dim_item", "untrimmed category", "in",
   i.filter(F.col("i_category") != F.trim(F.col("i_category"))).count())

dim_item = (dedup(i, ["i_item_sk"], "i_rec_start_date")
    .select(
        F.col("i_item_sk").cast("int").alias("item_sk"),
        F.col("i_item_id").alias("item_id"),
        F.trim(F.col("i_product_name")).alias("product_name"),
        F.trim(F.col("i_item_desc")).alias("product_description"),
        # trim the leading space the merch ERP feed injects into category
        F.upper(F.trim(F.col("i_category"))).alias("category"),
        F.col("i_category_id").cast("int").alias("category_id"),
        F.upper(F.trim(F.col("i_class"))).alias("product_class"),
        F.col("i_class_id").cast("int").alias("class_id"),
        F.trim(F.col("i_brand")).alias("brand"),
        F.col("i_brand_id").cast("int").alias("brand_id"),
        F.trim(F.col("i_manufact")).alias("manufacturer"),
        F.col("i_manager_id").cast("int").alias("manager_id"),
        F.trim(F.col("i_size")).alias("size"),
        F.trim(F.col("i_color")).alias("color"),
        F.trim(F.col("i_units")).alias("unit_of_measure"),
        F.trim(F.col("i_container")).alias("container"),
        F.col("i_current_price").cast("decimal(7,2)").alias("list_price"),
        F.col("i_wholesale_cost").cast("decimal(7,2)").alias("wholesale_cost"),
        F.col("i_rec_start_date").cast("date").alias("valid_from"),
        F.col("i_rec_end_date").cast("date").alias("valid_to"))
    .withColumn("is_current", F.col("valid_to").isNull())
    .withColumn("is_price_missing", F.col("list_price").isNull())
    .withColumn("unit_margin",
                F.when(F.col("list_price").isNotNull(),
                       (F.col("list_price") - F.col("wholesale_cost")).cast("decimal(9,2)")))
    .withColumn("margin_pct",
                F.when(F.col("list_price") > 0,
                       F.round((F.col("list_price") - F.col("wholesale_cost")) / F.col("list_price") * 100, 2)))
    .withColumn("price_tier",
        F.when(F.col("list_price").isNull(), "UNPRICED")
         .when(F.col("list_price") < 10, "VALUE")
         .when(F.col("list_price") < 50, "STANDARD")
         .when(F.col("list_price") < 150, "PREMIUM")
         .otherwise("LUXURY")))
n = publish(dim_item, "dim_item", files=1,
            comment="Product master with margin, price tier and SCD-2 validity window")
dq("dim_item", "rows published", "out", n)
dq("dim_item", "current records", "out",
   spark.table(f"{SILVER_DB}.dim_item").filter("is_current").count())
```

```python
from IPython.display import Markdown, display
display(Markdown(r"""
### 1.6 · The conforming step — one fact from three channels

The heart of the medallion pattern.

`store_sales`, `catalog_sales` and `web_sales` are three different tables, with three
different column vocabularies, at three different grains of detail. `conform()` projects
each onto **one canonical schema**, filling in a typed `NULL` where a channel genuinely
has no equivalent column — a catalogue order has no web page, a store sale has no
shipping address.

The result is a single `fct_sales` where *"revenue by channel"* is a `GROUP BY`, not a
three-way union written by hand in every query.
"""))
```

```python
# ---------------------------------------------------------------------
# 1.6  THE CONFORMING STEP — one sales fact from three channels
#
#      This is the single most valuable thing silver does. Store, catalog
#      and web arrive with different column names, different keys and
#      different measures. Downstream nobody should have to know that.
# ---------------------------------------------------------------------
section("1.6", "THE CONFORMING STEP — ONE FACT FROM THREE CHANNELS")
CANON_SALES = [
    ("sold_date_sk", "int"), ("sold_time_sk", "int"), ("ship_date_sk", "int"),
    ("item_sk", "int"), ("customer_sk", "int"), ("ship_customer_sk", "int"),
    ("address_sk", "int"), ("cdemo_sk", "int"), ("hdemo_sk", "int"),
    ("store_sk", "int"), ("call_center_sk", "int"), ("catalog_page_sk", "int"),
    ("web_site_sk", "int"), ("web_page_sk", "int"),
    ("ship_mode_sk", "int"), ("warehouse_sk", "int"), ("promo_sk", "int"),
    ("order_number", "bigint"), ("quantity", "int"),
    ("wholesale_cost", "decimal(7,2)"), ("list_price", "decimal(7,2)"),
    ("sales_price", "decimal(7,2)"), ("ext_discount_amt", "decimal(7,2)"),
    ("ext_sales_price", "decimal(7,2)"), ("ext_wholesale_cost", "decimal(7,2)"),
    ("ext_list_price", "decimal(7,2)"), ("ext_tax", "decimal(7,2)"),
    ("coupon_amt", "decimal(7,2)"), ("ext_ship_cost", "decimal(7,2)"),
    ("net_paid", "decimal(7,2)"), ("net_paid_inc_tax", "decimal(7,2)"),
    ("net_profit", "decimal(7,2)"),
]

SALES_MAP = {
 "STORE": ("store_sales", {
    "sold_date_sk":"ss_sold_date_sk", "sold_time_sk":"ss_sold_time_sk",
    "item_sk":"ss_item_sk", "customer_sk":"ss_customer_sk", "address_sk":"ss_addr_sk",
    "cdemo_sk":"ss_cdemo_sk", "hdemo_sk":"ss_hdemo_sk", "store_sk":"ss_store_sk",
    "promo_sk":"ss_promo_sk", "order_number":"ss_ticket_number", "quantity":"ss_quantity",
    "wholesale_cost":"ss_wholesale_cost", "list_price":"ss_list_price",
    "sales_price":"ss_sales_price", "ext_discount_amt":"ss_ext_discount_amt",
    "ext_sales_price":"ss_ext_sales_price", "ext_wholesale_cost":"ss_ext_wholesale_cost",
    "ext_list_price":"ss_ext_list_price", "ext_tax":"ss_ext_tax",
    "coupon_amt":"ss_coupon_amt", "net_paid":"ss_net_paid",
    "net_paid_inc_tax":"ss_net_paid_inc_tax", "net_profit":"ss_net_profit"}),
 "CATALOG": ("catalog_sales", {
    "sold_date_sk":"cs_sold_date_sk", "sold_time_sk":"cs_sold_time_sk",
    "ship_date_sk":"cs_ship_date_sk", "item_sk":"cs_item_sk",
    "customer_sk":"cs_bill_customer_sk", "ship_customer_sk":"cs_ship_customer_sk",
    "address_sk":"cs_bill_addr_sk", "cdemo_sk":"cs_bill_cdemo_sk", "hdemo_sk":"cs_bill_hdemo_sk",
    "call_center_sk":"cs_call_center_sk", "catalog_page_sk":"cs_catalog_page_sk",
    "ship_mode_sk":"cs_ship_mode_sk", "warehouse_sk":"cs_warehouse_sk",
    "promo_sk":"cs_promo_sk", "order_number":"cs_order_number", "quantity":"cs_quantity",
    "wholesale_cost":"cs_wholesale_cost", "list_price":"cs_list_price",
    "sales_price":"cs_sales_price", "ext_discount_amt":"cs_ext_discount_amt",
    "ext_sales_price":"cs_ext_sales_price", "ext_wholesale_cost":"cs_ext_wholesale_cost",
    "ext_list_price":"cs_ext_list_price", "ext_tax":"cs_ext_tax",
    "coupon_amt":"cs_coupon_amt", "ext_ship_cost":"cs_ext_ship_cost",
    "net_paid":"cs_net_paid", "net_paid_inc_tax":"cs_net_paid_inc_tax",
    "net_profit":"cs_net_profit"}),
 "WEB": ("web_sales", {
    "sold_date_sk":"ws_sold_date_sk", "sold_time_sk":"ws_sold_time_sk",
    "ship_date_sk":"ws_ship_date_sk", "item_sk":"ws_item_sk",
    "customer_sk":"ws_bill_customer_sk", "ship_customer_sk":"ws_ship_customer_sk",
    "address_sk":"ws_bill_addr_sk", "cdemo_sk":"ws_bill_cdemo_sk", "hdemo_sk":"ws_bill_hdemo_sk",
    "web_site_sk":"ws_web_site_sk", "web_page_sk":"ws_web_page_sk",
    "ship_mode_sk":"ws_ship_mode_sk", "warehouse_sk":"ws_warehouse_sk",
    "promo_sk":"ws_promo_sk", "order_number":"ws_order_number", "quantity":"ws_quantity",
    "wholesale_cost":"ws_wholesale_cost", "list_price":"ws_list_price",
    "sales_price":"ws_sales_price", "ext_discount_amt":"ws_ext_discount_amt",
    "ext_sales_price":"ws_ext_sales_price", "ext_wholesale_cost":"ws_ext_wholesale_cost",
    "ext_list_price":"ws_ext_list_price", "ext_tax":"ws_ext_tax",
    "coupon_amt":"ws_coupon_amt", "ext_ship_cost":"ws_ext_ship_cost",
    "net_paid":"ws_net_paid", "net_paid_inc_tax":"ws_net_paid_inc_tax",
    "net_profit":"ws_net_profit"}),
}


def conform(df, canon, mapping, channel):
    """Project a channel-specific table onto the canonical schema.
    A column the channel does not have becomes a typed NULL, not an error."""
    cols, missing = [], []
    for name, dtype in canon:
        src = mapping.get(name)
        if src and src in df.columns:
            cols.append(F.col(src).cast(dtype).alias(name))
        else:
            missing.append(name)
            cols.append(F.lit(None).cast(dtype).alias(name))
    if missing:
        print("      %-8s has no: %s" % (channel, ", ".join(missing)))
    return df.select(*cols).withColumn("channel", F.lit(channel))


raw_sales = None
for channel, (tbl, mapping) in SALES_MAP.items():
    part = conform(bronze(tbl), CANON_SALES, mapping, channel)
    raw_sales = part if raw_sales is None else raw_sales.unionByName(part)

# The union is read three times below (count, dedup, publish). Caching 51M
# rows would not fit the driver heap, so stage it to disk once instead —
# every later read is then a cheap columnar scan of one narrow table.
raw_sales = demo_slice(raw_sales, "channel", "order_number", "item_sk")
stage(raw_sales, "_stg_sales")
raw_sales = spark.table(f"{SILVER_DB}._stg_sales")
n_raw = raw_sales.count()
dq("fct_sales", "bronze rows (3 channels)", "in", n_raw)
```

```python
from IPython.display import Markdown, display
display(Markdown(r"""
### 1.7 · Deduplicate, quarantine, publish — then record the evidence

Three things happen here, in order:

1. **Dedup the CDC replays.** The natural key of a sales line is
   `(channel, order_number, item_sk)`. Bronze has duplicates on it because the change
   feed was replayed. A `row_number()` window keeps the last version — exactly what a
   MERGE would do.
2. **Quarantine, don't delete.** Rows failing a data contract — missing sold date,
   non-positive quantity, unsettled amount — are written to `dq_rejects_sales` *with
   the reason attached*. They are recoverable and auditable; a `WHERE` clause that
   dropped them would be neither.
3. **Publish** the clean fact, partitioned by year.

Finally, **§1.10 persists the data-quality ledger**. `dq_metrics` is the deliverable
most demos skip: every `dq ·` line printed above is also written to a table, so the
question *"how do you know the data is good?"* has a SQL answer rather than a
scrollback answer.
"""))
```

```python
# ---------------------------------------------------------------------
# 1.7  Dedup the CDC replays, quarantine the bad rows, publish fct_sales
# ---------------------------------------------------------------------
# The natural key of a sales line is (channel, order_number, item_sk).
# Bronze has duplicates on it because the CDC batch was replayed.
# Read three times below (the count, the quarantine and the fact), so stage it
# to disk once rather than recomputing the shuffle each time. On this heap a
# .cache() of even the sliced set is slower to re-read than plain Parquet.
section("1.7", "DEDUPLICATE, QUARANTINE, PUBLISH — THEN RECORD THE EVIDENCE")
stage(dedup(raw_sales, ["channel", "order_number", "item_sk"], "sold_date_sk"),
      "_stg_sales_dedup")
deduped = spark.table(f"{SILVER_DB}._stg_sales_dedup")
n_dedup = deduped.count()
dq("fct_sales", "duplicate lines removed", "clean", n_raw - n_dedup)

# The data contract: a sales line must have a date, an item, a positive
# quantity and a settled amount. Anything else is quarantined, with the
# reason attached, so operations can chase the source system.
reject_reason = (F.when(F.col("sold_date_sk").isNull(), "MISSING_SOLD_DATE")
                  .when(F.col("item_sk").isNull(), "MISSING_ITEM")
                  .when(F.col("net_paid").isNull(), "UNSETTLED_AMOUNT")
                  .when(F.col("quantity") <= 0, "NON_POSITIVE_QUANTITY"))

flagged = deduped.withColumn("reject_reason", reject_reason)
rejects = flagged.filter(F.col("reject_reason").isNotNull())
clean = flagged.filter(F.col("reject_reason").isNull()).drop("reject_reason")

publish(rejects.withColumn("rejected_at", F.current_timestamp()),
        "dq_rejects_sales", files=2,
        comment="Quarantined sales lines that failed the silver data contract")
for row in (spark.table(f"{SILVER_DB}.dq_rejects_sales")
                 .groupBy("reject_reason").count().collect()):
    dq("fct_sales", "rejected: " + row["reject_reason"], "reject", row["count"])

dim_date_lkp = spark.table(f"{SILVER_DB}.dim_date").select(
    F.col("date_sk").alias("_dsk"), F.col("calendar_date").alias("sold_date"),
    F.col("calendar_year").alias("sold_year"), F.col("month_of_year").alias("sold_month"),
    F.col("year_month").alias("sold_year_month"))

fct_sales = (clean
    .join(F.broadcast(dim_date_lkp), F.col("sold_date_sk") == F.col("_dsk"), "left").drop("_dsk")
    # one stable surrogate key for the line, independent of source system
    .withColumn("sales_line_sk",
                F.sha2(F.concat_ws("|", F.col("channel"), F.col("order_number"),
                                   F.col("item_sk")), 256).substr(1, 32))
    # business measures every downstream mart would otherwise recompute
    .withColumn("gross_revenue", F.col("ext_sales_price"))
    .withColumn("discount_amount",
                (F.coalesce(F.col("ext_discount_amt"), F.lit(0)) +
                 F.coalesce(F.col("coupon_amt"), F.lit(0))).cast("decimal(9,2)"))
    .withColumn("net_revenue", F.col("net_paid"))
    .withColumn("cost_of_goods", F.col("ext_wholesale_cost"))
    .withColumn("gross_margin",
                (F.col("net_paid") - F.coalesce(F.col("ext_wholesale_cost"), F.lit(0))).cast("decimal(9,2)"))
    .withColumn("margin_pct",
                F.when(F.col("net_paid") > 0,
                       F.round(F.col("gross_margin") / F.col("net_paid") * 100, 2)))
    .withColumn("discount_pct",
                F.when(F.col("ext_list_price") > 0,
                       F.round(F.col("discount_amount") / F.col("ext_list_price") * 100, 2)))
    .withColumn("is_promotional",
                F.col("promo_sk").isNotNull() & (F.col("discount_amount") > 0))
    .withColumn("is_shipped", F.col("ship_date_sk").isNotNull())
    .withColumn("days_to_ship",
                F.when(F.col("ship_date_sk").isNotNull(),
                       F.col("ship_date_sk") - F.col("sold_date_sk")))
    .filter(F.col("sold_year").isNotNull()))

n = publish(fct_sales, "fct_sales", partition_by="sold_year", files=8,
            comment="Conformed sales fact — store, catalog and web at one grain")
dq("fct_sales", "rows published", "out", n)
for _t in ("_stg_sales", "_stg_sales_dedup"):
    spark.sql(f"DROP TABLE IF EXISTS {SILVER_DB}.{_t}")

print()
spark.sql(f"""SELECT channel, sold_year, count(*) AS lines,
                     round(sum(net_revenue)/1e6, 1) AS net_revenue_musd
              FROM {SILVER_DB}.fct_sales GROUP BY channel, sold_year
              ORDER BY sold_year, channel""").show(30, False)

# %
# ---------------------------------------------------------------------
# 1.10  Persist this part's slice of the data-quality ledger
# ---------------------------------------------------------------------
schema = StructType([
    StructField("table_name", StringType()), StructField("rule", StringType()),
    StructField("scope", StringType()), StructField("metric_value", LongType()),
])
(spark.createDataFrame(DQ, schema)
      .withColumn("batch_id", F.lit(BATCH_ID))
      .withColumn("measured_at", F.current_timestamp())
      .repartition(1)
      .write.mode("overwrite").format("parquet")
      .saveAsTable(f"{SILVER_DB}.dq_metrics"))
print("dq_metrics rows written (overwrite):", len(DQ))

section("1.9", "SILVER TABLES SO FAR")
spark.sql(f"SHOW TABLES IN {SILVER_DB}").show(30, False)
finished(T_RUN, "SILVER PART A")
print("   next: 01b_silver_facts_returns_inventory")
```
