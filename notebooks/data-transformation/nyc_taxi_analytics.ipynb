```python
-- Daily revenue generated by taxi trips
SELECT
    DATE(tpep_pickup_datetime) AS trip_date,
    ROUND(SUM(total_amount), 2) AS total_revenue,
    COUNT(*) AS total_trips
FROM nyc_taxi.yellow_tripdata
GROUP BY DATE(tpep_pickup_datetime)
ORDER BY trip_date;
```

```python
-- Distribution of payment methods
SELECT
    payment_type,
    COUNT(*) AS trips,
    ROUND(SUM(total_amount), 2) AS revenue
FROM nyc_taxi.yellow_tripdata
GROUP BY payment_type
ORDER BY revenue DESC;
```

```python
-- Revenue generated from each pickup location
SELECT
    PULocationID,
    COUNT(*) AS trips,
    ROUND(SUM(total_amount), 2) AS revenue
FROM nyc_taxi.yellow_tripdata
GROUP BY PULocationID
ORDER BY revenue DESC
LIMIT 20;
```

```python
-- Trip volume by pickup hour
SELECT
    HOUR(tpep_pickup_datetime) AS pickup_hour,
    COUNT(*) AS total_trips
FROM nyc_taxi.yellow_tripdata
GROUP BY HOUR(tpep_pickup_datetime)
ORDER BY pickup_hour; 
```

```python
-- Trip volume by pickup hour
SELECT
    HOUR(tpep_pickup_datetime) AS pickup_hour,
    COUNT(*) AS total_trips
FROM nyc_taxi.yellow_tripdata
GROUP BY HOUR(tpep_pickup_datetime)
ORDER BY pickup_hour;
```

```python
-- Average trip distance by passenger count
SELECT
    passenger_count,
    ROUND(AVG(trip_distance), 2) AS avg_distance,
    COUNT(*) AS trips
FROM nyc_taxi.yellow_tripdata
GROUP BY passenger_count
ORDER BY passenger_count;
```

```python
-- Highest tips received
SELECT
    VendorID,
    trip_distance,
    fare_amount,
    tip_amount,
    total_amount
FROM nyc_taxi.yellow_tripdata
ORDER BY tip_amount DESC
LIMIT 20;
```

```python
-- Average fare across pickup hours
SELECT
    HOUR(tpep_pickup_datetime) AS hour_of_day,
    ROUND(AVG(fare_amount), 2) AS avg_fare
FROM nyc_taxi.yellow_tripdata
GROUP BY HOUR(tpep_pickup_datetime)
ORDER BY hour_of_day;
```

```python
-- Trips longer than 20 miles
SELECT
    VendorID,
    trip_distance,
    fare_amount,
    total_amount
FROM nyc_taxi.yellow_tripdata
WHERE trip_distance > 20
ORDER BY trip_distance DESC;
```
