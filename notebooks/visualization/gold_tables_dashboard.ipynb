# Dashboard: Gold Table Highlights

Reads each dataset's gold table and renders one polished Plotly chart per
dataset — a quick "here's what the demo data looks like" walkthrough.
Requires the corresponding generator + transformation notebooks to have
run first (see `notebooks/README.md` for the full list).

Requires `plotly` and `pandas` (see `requirements.txt` in this folder).

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
import plotly.express as px

spark = SparkSession.builder.getOrCreate()
TEMPLATE = "plotly_white"
```

## NYC Taxi — Daily Revenue

```python
taxi_pd = spark.table("nyc_taxi.gold_taxi_trip_summary_v1").orderBy("pickup_date").toPandas()

fig = px.line(
    taxi_pd, x="pickup_date", y="total_revenue", markers=True,
    title="NYC Taxi — Daily Revenue",
)
fig.update_layout(template=TEMPLATE, height=400)
fig.show()
```

## Citi Bike — Top Stations

```python
bike_pd = (
    spark.table("citibike.gold_station_activity_v1")
    .orderBy(F.desc("total_rides"))
    .limit(15)
    .toPandas()
    .sort_values("total_rides")
)

fig = px.bar(
    bike_pd, x="total_rides", y="start_station_name", orientation="h",
    color="total_rides", color_continuous_scale="Blues",
    title="Citi Bike — Top 15 Stations by Rides",
)
fig.update_layout(template=TEMPLATE, height=500, yaxis_title="", coloraxis_showscale=False)
fig.show()
```

## Wikipedia — Top Pages by Incoming Clicks

```python
wiki_pd = (
    spark.table("wikipedia.gold_clickstream_top_pages_v1")
    .orderBy(F.desc("total_clicks"))
    .limit(15)
    .toPandas()
    .sort_values("total_clicks")
)

fig = px.bar(
    wiki_pd, x="total_clicks", y="page", orientation="h",
    color="total_clicks", color_continuous_scale="Purples",
    title="Wikipedia — Top 15 Pages by Incoming Clicks",
)
fig.update_layout(template=TEMPLATE, height=500, yaxis_title="", coloraxis_showscale=False)
fig.show()
```

## Earthquakes — Daily Count & Avg Magnitude

```python
eq_pd = spark.table("earthquake_data.gold_earthquake_daily_summary_v1").orderBy("event_date").toPandas()

fig = px.bar(
    eq_pd, x="event_date", y="total_quakes", color="avg_magnitude",
    color_continuous_scale="OrRd",
    title="Earthquakes — Daily Count & Avg Magnitude",
)
fig.update_layout(template=TEMPLATE, height=400)
fig.show()
```

## Weather — Monthly Temperature Range

```python
wx_pd = spark.table("noaa_weather.gold_weather_monthly_summary_v1").orderBy("month").toPandas()

fig = px.line(
    wx_pd, x="month", y=["avg_temp_f", "max_temp_f", "min_temp_f"], markers=True,
    title="Weather — Monthly Temperature Range (Seattle Boeing Field)",
)
fig.update_layout(template=TEMPLATE, height=400, legend_title="")
fig.show()
```
